# CH01.3. **Cardinality of Sets**

> ## **동수(Equinumerous)**

#### **(1) 정의** : 두 집합 사이에 전단사 함수가 존재하는 관계
#### $ \Rrightarrow{} A\sim{}B $
#### $ \Leftrightarrow{} \exists{}f:A\rightarrow{}B \;\; \text{s.t. } \, f \text{ is bijective} $

#### **(2) 성질** :
##### $ \hspace{0.15cm} $ ① 반사성(Reflexivity) : $ \; A\sim{}A $
##### $ \hspace{0.15cm} $ ② 대칭성(Symmetry) : $ \; A\sim{}B \;\; \Rightarrow{} \;\; B\sim{}A $
##### $ \hspace{0.15cm} $ ③ 추이성(Transitivity) : $ \; A\sim{}B,\;\; B\sim{}C \;\; \Rightarrow{} \;\; A\sim{}C $

<b></b>

> ## **집합의 크기(Cardinality)**

#### **(1) 정의** : 집합에 포함된 원소의 수량적 크기를 나타내는 값
#### $ \Rrightarrow{} |A| $

##### **(`PLUS`)** 두 집합의 크기가 같다는 것은 두 집합이 동수라는 것과 동치임
##### $ \hspace{0.15cm} \Rrightarrow{} |A|=|B| \;\; \Leftrightarrow{} \;\; A\sim{}B $

<b></b>

> ## **기수의 비교(Cardinality Comparison)**

#### **(1) 정의** : 두 집합의 기수 사이의 동일성 또는 대소 관계

#### **(2) 비교 관계** :
##### $ \hspace{0.15cm} $ ① 동일 관계(Equality)
##### $ \hspace{0.45cm} |A|=|B| \;\; \Leftrightarrow{} \;\; \exists{}f:A\rightarrow{}B \;\; \text{s.t. } \, f \text{ is bijective} $
##### $ \hspace{0.15cm} $ ② 비엄격 순서 관계(Non-strict Order)
##### $ \hspace{0.45cm} |A|\leq{}|B| \;\; \Leftrightarrow{} \;\; \exists{}f:A\rightarrow{}B \;\; \text{s.t. } \, f \text{ is injective} $
##### $ \hspace{0.15cm} $ ③ 엄격 순서 관계(Strict Order)
##### $ \hspace{0.45cm} |A|<|B| \;\; \Leftrightarrow{} \;\; |A|\leq{}|B|,\;\;|A|\neq{}|B| $

##### **(`PLUS`)** 집합의 포함 관계는 기수의 비엄격 순서 관계를 보존함
##### $ \hspace{0.15cm} \Rrightarrow{} A\subseteq{}B \;\; \Rightarrow{} \;\; |A|\leq{}|B| $

##### **(`PLUS`)** 무한 집합에서는 진부분집합이 원래 집합과 동일한 기수를 가지는 것도 가능함
##### $ \hspace{0.15cm} \Rrightarrow{} \mathbb{N}\sim{}\{2,4,6,\cdots{}\} $

<b></b>

> ## (Theorem 1-7) **Cantor–Schröder–Bernstein Theorem**

#### 집합 $ A $에서 $ B $로 가는 단사 함수와 집합 $ B $에서 $ A $로 가는 단사 함수가 모두 존재하면 두 집합의 크기는 같다.
#### $ \Rrightarrow{} |A|\leq{}|B|,\;\; |B|\leq{}|A| \;\; \Rightarrow{} \;\; |A|=|B| $

<b></b>

> ## **유한 및 무한 집합(Finite and Infinite Set)**

#### **(0) 자연수 초기 구간(Natural Number Initial Segment)** : 처음 $ n $개의 자연수로 구성된 집합
#### $ \Rrightarrow{} [n] := \{1,2,\cdots{},n\} $
#### $ \hspace{0.45cm} [0] := \emptyset{} $

#### **(1) 유한 집합(Finite Set)** : 어떤 자연수 초기 구간 $ [n] $과 동수인 집합
#### $ \Rrightarrow{} A \text{ is finite} $
#### $ \Leftrightarrow{} \exists{}n\in{}\mathbb{N}_{0} \;\; \text{s.t. } \, A\sim{}[n] $
#### $ \hspace{0.45cm} \text{where } \, \mathbb{N}_{0}:=\{0,1,2,\cdots{}\} $

##### **(`PLUS`)** 유한 집합 $ A $가 $ [n] $과 동수이면 그 집합의 크기는 $ n $임
##### $ \hspace{0.15cm} \Rrightarrow{} A\sim{}[n] \;\; \Rightarrow{} \;\; |A|=n $

#### **(3) 무한 집합(Infinite Set)** : 어떠한 자연수 초기 구간과도 동수가 아닌 집합
#### $ \Rrightarrow{} A \text{ is infinite} $
#### $ \Leftrightarrow{} \forall{}n\in{}\mathbb{N}_{0},\;\; A\not\sim{}[n] $

<b></b>

> ## (Theorem 1-8) **Inclusion-Exclusion Principle**

#### 유한 집합들의 합집합의 크기는 각 집합의 크기에서 중복하여 계산된 교집합의 크기를 보정하여 계산할 수 있다.

#### **(1)** 두 유한 집합의 합집합의 크기는 각 집합의 크기의 합에서 두 집합의 교집합의 크기를 뺀 값과 같다.
#### $ \Rrightarrow{} |A\cup{}B| = |A|+|B|-|A\cap{}B| $

#### **(2)** 세 유한 집합의 합집합의 크기는 각 집합과 교집합의 크기를 포함-배제하여 계산한다.
#### $ \Rrightarrow{} |A\cup{}B\cup{}C| $
#### $ \hspace{0.45cm} = |A|+|B|+|C| $
#### $ \hspace{0.65cm} -|A\cap{}B|-|A\cap{}C|-|B\cap{}C| $
#### $ \hspace{0.65cm} +|A\cap{}B\cap{}C| $

<b></b>

> ## **가산 및 불가산 집합(Countable and Uncountable Sets)**

#### **(1) 가산 집합(Countable Set)** : 자연수 집합 $ \mathbb{N} $으로 가는 단사 함수가 존재하는 집합
#### $ \Rrightarrow{} A \text{ is countable} $
#### $ \Leftrightarrow{} \exists{}f:A\rightarrow{}\mathbb{N} \;\; \text{s.t. } \, f \text{ is injective} $

##### **(`PLUS`)** 해당 노트에서 가산 집합은 유한 집합과 가산-무한 집합을 모두 포함하는 개념으로 사용함

#### **(2) 가산-무한 집합(Countably Infinite Set)** : 자연수 집합 $ \mathbb{N} $과 동수인 집합
#### $ \Rrightarrow{} A \text{ is countably infinite} $
#### $ \Leftrightarrow{} A\sim{}\mathbb{N} $

#### **(3) 불가산 집합(Uncountable Set)** : 가산 집합이 아닌 집합
#### $ \Rrightarrow{} A \text{ is uncountable} $
#### $ \Leftrightarrow{} \nexists{}f:A\rightarrow{}\mathbb{N} \;\; \text{s.t. } \, f \text{ is injective} $

#### **(4) 예시** :
##### $ \hspace{0.15cm} $ ① 자연수 집합, 정수 집합, 유리수 집합은 모두 가산-무한 집합임
##### $ \hspace{0.45cm} |\mathbb{N}|=|\mathbb{Z}|=|\mathbb{Q}| $
##### $ \hspace{0.15cm} $ ② 실수 집합은 불가산 집합임
##### $ \hspace{0.45cm} |\mathbb{N}|<|\mathbb{R}| $

> ## (Theorem 1-9) **Basic Properties of Countability**

#### **(1)** 가산 집합의 모든 부분집합은 가산 집합이다.
#### $ \Rrightarrow{} B \text{ is countable},\;\; A\subseteq{}B \;\; \Rightarrow{} \;\; A \text{ is countable} $

#### **(2)** 불가산 집합을 포함하는 모든 초집합은 불가산 집합이다.
#### $ \Rrightarrow{} A \text{ is uncountable},\;\; A\subseteq{}B \;\; \Rightarrow{} \;\; B \text{ is uncountable} $

#### **(3)** 실수의 모든 비퇴화 유계 구간은 불가산 집합이다.
#### $ \Rrightarrow{} a<b \;\; \Rightarrow{} \;\; (a,b),\;[a,b],\;[a,b),\;(a,b] \text{ are uncountable} $

<b></b>

> ## (Theorem 1-10) **Cartesian Product of Countable Sets**

#### 두 집합 $ A $, $ B $가 가산 집합이면 곱집합 $ A\times{}B $도 가산 집합이다.
#### $ \Rrightarrow{} A,B \text{ are countable} \;\; \Rightarrow{} \;\; A\times{}B \text{ is countable} $

<b></b>

> ## (Theorem 1-11) **Cantor's Theorem**

#### **(1)** 집합 $ A $와 그 멱집합 $ 2^{A} $ 사이에는 전단사 함수가 존재하지 않는다.
#### $ \Rrightarrow{} |A|\neq{}|2^{A}| $

#### **(2)** 집합 $ A $에서 그 멱집합 $ 2^{A} $으로 가는 단사 함수는 존재하므로 멱집합의 크기가 원래 집합보다 크다.
#### $ \Rrightarrow{} |A|<|2^{A}| $

##### **(`PLUS`)** 유한 집합 $ A $의 원소 개수가 $ n $개이면 멱집합은 $ 2^{n} $개의 원소를 가짐
##### $ \hspace{0.15cm} \Rrightarrow{} |A|=n \;\; \Rightarrow{} \;\; |2^{A}|=2^{n} $
