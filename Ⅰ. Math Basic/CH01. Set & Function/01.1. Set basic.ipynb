# CH01.1. **Set basic**

> ## **집합(Set)**

#### **(1) 정의** : 중복과 순서를 고려하지 않는 원소(Element)들의 모음
#### $ \Rrightarrow{} A = \{x \mid{} x \text{ satisfies a given condition}\} $

#### **(2) 원소 포함 관계** : 원소 $ x $가 집합 $ A $에 포함되는지 여부를 다음과 같이 표현함
##### $ \hspace{0.15cm} $ ① 원소 $ x $가 집합 $ A $에 속함
##### $ \hspace{0.45cm} x \in{} A $
##### $ \hspace{0.15cm} $ ② 원소 $ x $가 집합 $ A $에 속하지 않음
##### $ \hspace{0.45cm} x \notin{} A $

#### **(3) 수의 체계 표현** :
##### $ \hspace{0.15cm} $ ① 자연수(Natural Number)
##### $ \hspace{0.45cm} \mathbb{N} := \{1,2,3,\cdots{}\} $

##### **(`PLUS`)** 해당 노트에서는 자연수의 정의에서는 $ 0 $을 포함하지 않음

##### $ \hspace{0.15cm} $ ② 정수(Integer)
##### $ \hspace{0.45cm} \mathbb{Z} := \{\cdots{},-2,-1,0,1,2,\cdots{}\} $
##### $ \hspace{0.15cm} $ ③ 유리수(Rational Number)
##### $ \hspace{0.45cm} \mathbb{Q} := \left\{\frac{p}{q} : p,q \in{} \mathbb{Z},\;\; q \neq{} 0\right\} $
##### $ \hspace{0.15cm} $ ④ 실수(Real Number)
##### $ \hspace{0.45cm} \mathbb{R} := \text{유리수와 무리수를 모두 포함하는 수의 집합} $

#### **(4) 수의 범위 표현** :
##### $ \hspace{0.15cm} $ ① 열린 구간(Open Interval)
##### $ \hspace{0.45cm} (a,b) := \{x \in{} \mathbb{R} : a < x < b\} $
##### $ \hspace{0.15cm} $ ② 닫힌 구간(Closed Interval)
##### $ \hspace{0.45cm} [a,b] := \{x \in{} \mathbb{R} : a \leq{} x \leq{} b\} $
##### $ \hspace{0.15cm} $ ③ 반열린 구간(Half-Open Interval)
##### $ \hspace{0.45cm} [a,b) := \{x \in{} \mathbb{R} : a \leq{} x < b\},\;\; (a,b] := \{x \in{} \mathbb{R} : a < x \leq{} b\} $
##### $ \hspace{0.15cm} $ ④ 무한 구간(Unbounded Interval)
##### $ \hspace{0.45cm} (-\infty{},a),\;\; (-\infty{},a],\;\; (a,\infty{}),\;\; [a,\infty{}) $
##### $ \hspace{0.15cm} $ ⑤ 퇴화 구간(Degenerate Interval)
##### $ \hspace{0.45cm} [a,a] = \{a\} $

#### **(5) 예시** :
##### $ \hspace{0.15cm} $ ① $ \{ 1, 2, 3, 4 \} $
##### $ \hspace{0.15cm} $ ② $ \{ 😀, 😡 \} = \{ 😡, 😀 \} $
##### $ \hspace{0.15cm} $ ③ $ \{ $가$ , $나$ , $다$ \} $

<b></b>

> ## **우주 집합과 공집합(Universal Set and Empty Set)**

#### **(1) 우주 집합(Universal Set)** : 현재 문맥에서 고려하는 모든 원소를 포함하는 전체 집합
#### $ \Rrightarrow{} U $

#### **(2) 공집합(Empty Set)** : 원소를 하나도 가지지 않는 집합
#### $ \Rrightarrow{} \emptyset{} := \{\} $

##### **(`PLUS`)** 공집합은 모든 집합의 부분집합임
##### $ \hspace{0.15cm} \Rrightarrow{} \emptyset{} \subseteq{} A $

<b></b>

> ## **부분집합(Subset)**

#### **(1) 정의** : 집합 $ A $의 모든 원소가 집합 $ B $에도 포함되는 관계
#### $ \Rrightarrow{} A \subseteq{} B $
#### $ \Leftrightarrow{} (\forall{}x)\;\; x \in{} A \;\; \Rightarrow{} \;\; x \in{} B $

##### **(`PLUS`)** $ A \subseteq{} B $인 경우 집합 $ B $를 집합 $ A $의 초집합(Superset)으로 표현함

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ \{a,b\} \subseteq{} \{a,b,c\} $
##### $ \hspace{0.15cm} $ ② $ \{a,b,c\} \subseteq{} \{a,b,c\} $
##### $ \hspace{0.15cm} $ ③ $ \{1,4\} \not\subseteq{} \{1,2,3\} $

<b></b>

> ## **진부분집합(Proper Subset)**

#### **(1) 정의** : 집합 $ A $가 집합 $ B $의 부분집합이면서 두 집합이 서로 다른 관계
#### $ \Rrightarrow{} A \subsetneqq{} B $
#### $ \Leftrightarrow{} A \subseteq{} B,\;\; A \neq{} B $

<b></b>

> ## **멱집합(Power Set)**

#### **(1) 정의** : 집합 $ U $의 모든 부분집합을 원소로 가지는 집합
#### $ \Rrightarrow{} 2^{U} := \{A \mid{} A \subseteq{} U\} $

##### **(`PLUS`)** 멱집합은 $ \mathcal{P}(U) $로도 표현함
##### $ \hspace{0.15cm} \Rrightarrow{} \mathcal{P}(U) = 2^{U} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ U=\{1,2\} $인 경우
##### $ \hspace{0.45cm} 2^{U} = \{\emptyset{},\{1\},\{2\},\{1,2\}\} $
##### $ \hspace{0.15cm} $ ② $ U=\mathbb{R} $인 경우
##### $ \hspace{0.45cm} 2^{U} = \{A \mid{} A \subseteq{} \mathbb{R}\} $

<b></b>

> ## (Theorem 1-1) **Extensionality of Sets**

#### **(1)** 두 집합이 같을 필요충분조건은 모든 원소에 대해 두 집합의 원소 포함 여부가 동일한 것이다.
#### $ \Rrightarrow{} A=B \;\; \Leftrightarrow{} \;\; (\forall{}x)\;\; x\in{}A \;\; \Leftrightarrow{} \;\; x\in{}B $\

#### **(2)** 두 집합이 같을 필요충분조건은 두 집합이 서로의 부분집합인 것이다.
#### $ \Rrightarrow{} A=B \;\; \Leftrightarrow{} \;\; A\subseteq{}B \;\; \text{and} \;\; B\subseteq{}A $

#### **(2) 종류** :
##### $ \hspace{0.15cm} $ ① 합집합(Union) : 두 집합 중 적어도 하나에 속하는 원소들로 구성된 집합

> ## **서로소 집합(Disjoint Sets)**

#### **(1) 정의** : 두 집합이 공통된 원소를 가지지 않는 관계
#### $ \Rrightarrow{} A\cap{}B = \emptyset{} $

<b></b>

> ## **(Theorem 1-2) Basic Laws of Set Operations**

#### **(1)** 공집합은 합집합의 항등원이고 우주 집합은 교집합의 항등원이다.
#### $ \Rrightarrow{} A\cup{}\emptyset{} = A,\;\; A\cap{}U = A $

#### **(2)** 집합과 공집합의 교집합은 공집합이고 집합과 우주 집합의 합집합은 우주 집합이다.
#### $ \Rrightarrow{} A\cap{}\emptyset{} = \emptyset{},\;\; A\cup{}U = U $

#### **(3)** 동일한 집합끼리 합집합 또는 교집합을 취하면 원래 집합과 같다.
#### $ \Rrightarrow{} A\cup{}A = A,\;\; A\cap{}A = A $

#### **(4)** 집합과 그 여집합의 합집합은 우주 집합이고 교집합은 공집합이며, 여집합을 두 번 취하면 원래 집합과 같다.
#### $ \Rrightarrow{} A\cup{}A^{c}=U,\;\; A\cap{}A^{c}=\emptyset{},\;\; (A^{c})^{c}=A $

#### **(5)** 합집합과 교집합은 연산 순서를 바꾸어도 결과가 같다.
#### $ \Rrightarrow{} A\cup{}B = B\cup{}A,\;\; A\cap{}B = B\cap{}A $

#### **(6)** 합집합과 교집합은 세 집합의 연산 결합 순서를 바꾸어도 결과가 같다.
#### $ \Rrightarrow{} (A\cup{}B)\cup{}C = A\cup{}(B\cup{}C) $
#### $ \hspace{0.45cm} (A\cap{}B)\cap{}C = A\cap{}(B\cap{}C) $

<b></b>

> ## (Theorem 1-3) **De Morgan's Law**

#### **(1)** 두 집합의 합집합에 여집합을 취하면 각 집합의 여집합의 교집합과 같다.
#### $ \Rrightarrow{} (A\cup{}B)^{c} = A^{c}\cap{}B^{c} $

#### **(2)** 두 집합의 교집합에 여집합을 취하면 각 집합의 여집합의 합집합과 같다.
#### $ \Rrightarrow{} (A\cap{}B)^{c} = A^{c}\cup{}B^{c} $

> ## (Theorem 1-4) **Distributive Law of Sets**

#### 합집합과 교집합은 서로에 대해 분배 법칙을 만족한다.

#### **(1)** 교집합은 합집합에 대해 분배 법칙을 만족한다.
#### $ \Rrightarrow{} A\cap{}(B\cup{}C) = (A\cap{}B)\cup{}(A\cap{}C) $

#### **(2)** 합집합은 교집합에 대해 분배 법칙을 만족한다.
#### $ \Rrightarrow{} A\cup{}(B\cap{}C) = (A\cup{}B)\cap{}(A\cup{}C) $

<b></b>

> ## **순서쌍과 순서 튜플(Ordered Pair and Ordered Tuple)**

#### **(1) 순서쌍(Ordered Pair)** : 두 원소를 순서를 고려하여 나열한 객체
#### $ \Rrightarrow{} (a,b) $

##### **(`PLUS`)** 서로 다른 두 원소의 순서를 바꾸면 일반적으로 서로 다른 순서쌍이 됨
##### $ \hspace{0.15cm} \Rrightarrow{} (a,b)\neq{}(b,a) \;\; \text{if } \, a\neq{}b $

#### **(2) 순서 튜플(Ordered Tuple)** : $ n $개의 원소를 순서를 고려하여 나열한 객체
#### $ \Rrightarrow{} (a_{1},a_{2},\cdots{},a_{n}) $

<b></b>

> ## **곱집합(Cartesian Product)**

#### **(1) 정의** : 각 집합에서 하나의 원소를 선택하여 구성할 수 있는 모든 순서 튜플(Ordered Tuple)의 집합
#### $ \Rrightarrow{} A_{1}\times{}A_{2}\times{}\cdots{}\times{}A_{n} := \{(a_{1},a_{2},\cdots{},a_{n}) \mid{} a_{i}\in{}A_{i},\;\; i=1,2,\cdots{},n\} $

##### **(`PLUS`)** 두 집합 $ A $, $ B $의 곱집합은 모든 순서쌍 $ (a,b) $로 구성됨
##### $ \hspace{0.15cm} \Rrightarrow{} A\times{}B = \{(a,b) \mid{} a\in{}A,\;\; b\in{}B\} $

##### **(`PLUS`)** 동일한 집합 $ A $의 $ n $중 곱집합을 $ A^{n} $으로 표현함
##### $ \hspace{0.15cm} \Rrightarrow{} A^{n} := \underbrace{A\times{}A\times{}\cdots{}\times{}A}_{n\text{ times}} $

#### **(2) 성질** : 
##### $ \hspace{0.15cm} $ ① 곱집합은 원소의 순서를 고려함
##### $ \hspace{0.45cm} \Rrightarrow{} (1,a) \neq{} (a,1), \;\; A \times{} B \neq{} B \times{} A \;\; (\text{generally}) $

#### **(3) 예시** :
##### $ \hspace{0.15cm} $ ① 집합 $ A=\{1,2\} $, $ B=\{a,b,c\} $가 존재할 때
##### $ \hspace{0.45cm} A\times{}B = \{(1,a),(1,b),(1,c),(2,a),(2,b),(2,c)\} $
##### $ \hspace{0.45cm} B\times{}A = \{(a,1),(a,2),(b,1),(b,2),(c,1),(c,2)\} $
##### $ \hspace{0.15cm} $ ② 실수 집합의 $ 3 $중 곱집합
##### $ \hspace{0.45cm} \mathbb{R}^{3} = \mathbb{R}\times{}\mathbb{R}\times{}\mathbb{R} = \{(x,y,z) \mid{} x,y,z\in{}\mathbb{R}\} $
