# CH01.4. **Families and Sequences of Sets**

> ## **인덱스된 족(Indexed Family)**

#### **(1) 정의** : 인덱스 집합 $ I $의 각 원소에 집합 $ X $의 원소를 하나씩 대응시켜 구성한 객체들의 모음
#### $ \Rrightarrow{} x:I\rightarrow{}X,\;\; i\mapsto{}x_{i} $
#### $ \Rrightarrow{} (x_{i})_{i\in{}I} $
#### $ \hspace{0.45cm} \text{where } \, I:=\text{Index Set} $

##### **(`PLUS`)** 인덱스 집합은 자연수에 한정되지 않으며 유한 집합, 정수 집합, 실수 집합 등의 임의의 집합을 사용할 수 있음

##### **(`PLUS`)** 서로 다른 인덱스가 동일한 객체에 대응되는 것도 가능함
#### $ \hspace{0.15cm} \Rrightarrow{} i\neq{}j \;\; \nRightarrow{} \;\; x_{i}\neq{}x_{j} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① 수로 구성된 인덱스된 족
##### $ \hspace{0.45cm} x:I\rightarrow{}\mathbb{R},\;\; i\mapsto{}x_{i} $
##### $ \hspace{0.15cm} $ ② 집합으로 구성된 인덱스된 족
##### $ \hspace{0.45cm} A:I\rightarrow{}2^{U},\;\; i\mapsto{}A_{i} $
##### $ \hspace{0.15cm} $ ③ 함수로 구성된 인덱스된 족
##### $ \hspace{0.45cm} i\mapsto{}f_{i} $

<b></b>

> ## **집합족의 구조(Set Family Structure)**

#### **(1) 집합족(Family of Sets)** : 집합 $ U $의 부분집합들을 원소로 가지는 집합
#### $ \Rrightarrow{} \mathcal{C}\subseteq{}2^{U} $

#### **(2) 인덱스된 집합족(Indexed Family of Sets)** : 각 인덱스에 집합 $ U $의 부분집합 하나를 대응시킨 인덱스된 족
#### $ \Rrightarrow{} A:I\rightarrow{}2^{U},\;\; i\mapsto{}A_{i} $
#### $ \Rrightarrow{} (A_{i})_{i\in{}I} $

#### **(3) 연산** :
##### $ \hspace{0.15cm} $ ① 합집합(Union) : 적어도 하나의 집합 $ A_{i} $에 포함되는 원소들로 구성된 집합
##### $ \hspace{0.45cm} \bigcup_{i\in{}I}A_{i} := \{x\in{}U\mid{}\exists{}i\in{}I,\;\;x\in{}A_{i}\} $
##### $ \hspace{0.15cm} $ ② 교집합(Intersection) : 모든 집합 $ A_{i} $에 공통으로 포함되는 원소들로 구성된 집합
##### $ \hspace{0.45cm} \bigcap_{i\in{}I}A_{i} := \{x\in{}U\mid{}\forall{}i\in{}I,\;\;x\in{}A_{i}\} $

#### **(4) 성질** :
##### $ \hspace{0.15cm} $ ① 쌍별 서로소(Pairwise Disjoint) : 서로 다른 인덱스에 대응되는 모든 두 집합이 서로소인 성질
##### $ \hspace{0.45cm} A_{i}\cap{}A_{j}=\emptyset{} \;\; \forall{}i,j\in{}I,\;\;i\neq{}j $
##### $ \hspace{0.15cm} $ ② 덮개(Covering) : 집합족의 합집합이 전체 집합 $ U $와 동일한 성질
##### $ \hspace{0.45cm} \bigcup_{i\in{}I}A_{i}=U $
##### $ \hspace{0.15cm} $ ③ 중첩(Nested) : 임의의 두 집합이 포함 관계에 의해 서로 비교 가능한 성질
##### $ \hspace{0.45cm} A_{i}\subseteq{}A_{j} \;\; \text{or} \;\; A_{j}\subseteq{}A_{i} \;\; \forall{}i,j\in{}I $

<b></b>

> ## (Theorem 1-11) **Generalized De Morgan's Law**

#### **(1)** 합집합의 여집합은 각 집합의 여집합들의 교집합과 같다.
#### $ \Rrightarrow{} \left(\bigcup_{i\in{}I}A_{i}\right)^{c}=\bigcap_{i\in{}I}A_{i}^{c} $

#### **(2)** 교집합의 여집합은 각 집합의 여집합들의 합집합과 같다.
#### $ \Rrightarrow{} \left(\bigcap_{i\in{}I}A_{i}\right)^{c}=\bigcup_{i\in{}I}A_{i}^{c} $

<b></b>

> ## **집합의 분할(Partition of a Set)**

#### **(1) 정의** : 집합 $ A $를 서로 겹치지 않는 비어 있지 않은 부분집합들로 완전히 나누는 집합족
#### $ \Rrightarrow{} \mathcal{P}\subseteq{}2^{A} $

#### **(2) 조건** :
##### $ \hspace{0.15cm} $ ① 분할을 구성하는 각 집합은 공집합이 아님
##### $ \hspace{0.45cm} P\neq{}\emptyset{} \;\;\; \forall{}P\in{}\mathcal{P} $
##### $ \hspace{0.15cm} $ ② 서로 다른 두 집합은 서로소임
##### $ \hspace{0.45cm} P\cap{}Q=\emptyset{} \;\;\; \forall{}P,Q\in{}\mathcal{P},\;\;P\neq{}Q $
##### $ \hspace{0.15cm} $ ③ 모든 집합의 합집합은 원래 집합과 동일함
##### $ \hspace{0.45cm} \bigcup{}\mathcal{P}=A $

<b></b>

> ## (Theorem 1-12) **Equivalence Relations and Partitions**

#### **(1)** 집합 $ A $ 위의 동치 관계 $ R $에 의해 생성되는 서로 다른 동치류들은 $ A $의 분할을 이룬다.
#### $ \Rrightarrow{} A/R := \{[a]_{R}\mid{}a\in{}A\} $

#### **(2)** 집합 $ A $의 분할 $ \mathcal{P} $가 주어지면 동일한 분할 원소에 속하는 두 원소를 연결하여 동치 관계를 정의할 수 있다.
#### $ \Rrightarrow{} R_{\mathcal{P}} := \{(x,y)\in{}A\times{}A\mid{}\exists{}P\in{}\mathcal{P},\;\;x,y\in{}P\} $

<b></b>

> ## **열(Sequence)**

#### **(1) 정의** : 자연수 집합 $ \mathbb{N} $을 인덱스 집합으로 가지는 인덱스된 족
#### $ \Rrightarrow{} x:\mathbb{N}\rightarrow{}X,\;\; n\mapsto{}x_{n} $
#### $ \Rrightarrow{} (x_{n})_{n\in{}\mathbb{N}} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① 실수열(Sequence of Real Numbers) : $ \; x:\mathbb{N}\rightarrow{}\mathbb{R} $
##### $ \hspace{0.15cm} $ ② 함수열(Sequence of Functions) : $ \; (f_{n})_{n\in{}\mathbb{N}} $
##### $ \hspace{0.15cm} $ ③ 집합열(Set Sequence) : $ \; A:\mathbb{N}\rightarrow{}2^{U} $

<b></b>

> ## **집합열(Set Sequence)**

#### **(1) 정의** : 각 항이 집합으로 구성된 열
#### $ \Rrightarrow{} A:\mathbb{N}\rightarrow{}2^{U},\;\; n\mapsto{}A_{n} $
#### $ \Rrightarrow{} (A_{n})_{n\in{}\mathbb{N}} $

<b></b>

> ## (Theorem 1-13) **Countable Union of Countable Sets**

#### 모든 $ A_{n} $이 가산 집합이면 이들의 가산 합집합도 가산 집합이다.
#### $ \Rrightarrow{} \displaystyle{} A_{n}\text{ is countable }\;\;\forall{}n\in{}\mathbb{N} \;\; \Rightarrow{} \;\; \bigcup_{n=1}^{\infty{}}A_{n}\text{ is countable} $

<b></b>

> ## **단조 집합열(Monotone Set Sequence)**

#### **(1) 단조 증가 집합열(Increasing Set Sequence)** : 각 집합이 다음 집합의 부분집합인 집합열
#### $ \Rrightarrow{} A_{n}\subseteq{}A_{n+1} \;\;\; \forall{}n\in{}\mathbb{N} $
#### $ \Rrightarrow{} A_{1}\subseteq{}A_{2}\subseteq{}\cdots{} $

#### **(2) 단조 감소 집합열(Decreasing Set Sequence)** : 각 집합이 다음 집합을 포함하는 집합열
#### $ \Rrightarrow{} A_{n}\supseteq{}A_{n+1} \;\;\; \forall{}n\in{}\mathbb{N} $
#### $ \Rrightarrow{} A_{1}\supseteq{}A_{2}\supseteq{}\cdots{} $

<b></b>

> ## **집합열의 극한(Limit of Set Sequence)**

#### **(1) 상극한(Limit Superior)** : 집합열에서 무한히 반복하여 포함되는 원소들로 구성된 집합
#### $ \Rrightarrow{} \displaystyle{} \limsup_{n\rightarrow{}\infty{}}A_{n} := \bigcap_{n=1}^{\infty{}}\bigcup_{k\geq{}n}A_{k} $

##### **(`PLUS`)** 원소 $ x $가 상극한에 포함되는 것은 임의의 시점 이후에도 $ x $를 포함하는 집합이 다시 존재함을 의미함
##### $ \hspace{0.15cm} \Rrightarrow{} x\in{}\limsup_{n\rightarrow{}\infty{}}A_{n} $
##### $ \hspace{0.45cm} \Leftrightarrow{} \forall{}m\in{}\mathbb{N},\;\;\exists{}n\geq{}m \;\; \text{s.t. } \, x\in{}A_{n} $

#### **(2) 하극한(Limit Inferior)** : 어떤 시점 이후부터 계속 포함되는 원소들로 구성된 집합
#### $ \Rrightarrow{} \displaystyle{} \liminf_{n\rightarrow{}\infty{}}A_{n} := \bigcup_{n=1}^{\infty{}}\bigcap_{k\geq{}n}A_{k} $

##### **(`PLUS`)** 원소 $ x $가 하극한에 포함되는 것은 어떤 시점 이후의 모든 집합이 $ x $를 포함함을 의미함
##### $ \hspace{0.15cm} \Rrightarrow{} x\in{}\liminf_{n\rightarrow{}\infty{}}A_{n} $
##### $ \hspace{0.45cm} \Leftrightarrow{} \exists{}m\in{}\mathbb{N} \;\; \text{s.t. } \, x\in{}A_{n}\;\;\forall{}n\geq{}m $

#### **(3) 수렴(Convergence)** : 상극한과 하극한이 동일한 집합으로 일치하는 집합열
#### $ \Rrightarrow{} \displaystyle{} \liminf_{n\rightarrow{}\infty{}}A_{n}=\limsup_{n\rightarrow{}\infty{}}A_{n}=A $
#### $ \Rrightarrow{} \displaystyle{} \lim_{n\rightarrow{}\infty{}}A_{n}:=A $

<b></b>

> ## (Theorem 1-14) **Basic Properties of Set Limits**

#### **(1)** 모든 집합열의 하극한은 상극한의 부분집합이다.
#### $ \Rrightarrow{} \displaystyle{} \liminf_{n\rightarrow{}\infty{}}A_{n}\subseteq{}\limsup_{n\rightarrow{}\infty{}}A_{n} $

#### **(2)** 상극한과 하극한은 여집합 연산에 의해 서로 교환된다.
#### $ \Rrightarrow{} \displaystyle{} \left(\limsup_{n\rightarrow{}\infty{}}A_{n}\right)^{c}=\liminf_{n\rightarrow{}\infty{}}A_{n}^{c} $
#### $ \hspace{0.45cm} \displaystyle{} \left(\liminf_{n\rightarrow{}\infty{}}A_{n}\right)^{c}=\limsup_{n\rightarrow{}\infty{}}A_{n}^{c} $

#### **(3)** 단조 증가 집합열의 극한은 모든 집합의 합집합과 같다.
#### $ \Rrightarrow{} \displaystyle{} A_{1}\subseteq{}A_{2}\subseteq{}\cdots{} \;\; \Rightarrow{} \;\; \lim_{n\rightarrow{}\infty{}}A_{n}=\bigcup_{n=1}^{\infty{}}A_{n} $

#### **(4)** 단조 감소 집합열의 극한은 모든 집합의 교집합과 같다.
#### $ \Rrightarrow{} \displaystyle{} A_{1}\supseteq{}A_{2}\supseteq{}\cdots{} \;\; \Rightarrow{} \;\; \lim_{n\rightarrow{}\infty{}}A_{n}=\bigcap_{n=1}^{\infty{}}A_{n} $
