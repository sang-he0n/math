# CH01.2. **Functions**

> ## **이항 관계(Binary Relation)**

#### **(1) 정의** : 두 집합 $ A $, $ B $의 곱집합 $ A\times{}B $의 부분집합
#### $ \Rrightarrow{} R\subseteq{}A\times{}B $

##### **(`PLUS`)** 관계 $ R $의 원소는 두 집합의 원소로 구성된 순서쌍이며, $ (a,b)\in{}R $은 원소 $ a $, $ b $가 관계 $ R $에 의해 연결되어 있음을 의미함

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① 집합 $ A=\{1,2,3\} $ 위의 대소 관계
##### $ \hspace{0.45cm} R := \{(a,b)\in{}A\times{}A \mid{} a\leq{}b\} $
##### $ \hspace{0.15cm} $ ② 실수 집합 $ \mathbb{R} $ 위의 등호 관계
##### $ \hspace{0.45cm} R := \{(x,y)\in{}\mathbb{R}\times{}\mathbb{R} \mid{} x=y\} $

<b></b>

> ## **동치 관계(Equivalence Relation)**

#### **(1) 정의** : 집합 $ A $ 위의 이항 관계 중 반사성, 대칭성, 추이성을 모두 만족하는 관계
#### $ \Rrightarrow{} R\subseteq{}A\times{}A $

#### **(2) 조건** :
##### $ \hspace{0.15cm} $ ① 반사성(Reflexivity) : 모든 원소가 자기 자신과 관계를 가짐
##### $ \hspace{0.45cm} (a,a)\in{}R \;\;\; \forall{}a\in{}A $
##### $ \hspace{0.15cm} $ ② 대칭성(Symmetry) : 두 원소 사이의 관계가 반대 방향에서도 성립함
##### $ \hspace{0.45cm} (a,b)\in{}R \;\; \Rightarrow{} \;\; (b,a)\in{}R $
##### $ \hspace{0.15cm} $ ③ 추이성(Transitivity) : 연속된 두 관계가 성립하면 처음과 마지막 원소 사이의 관계도 성립함
##### $ \hspace{0.45cm} (a,b)\in{}R,\;\; (b,c)\in{}R \;\; \Rightarrow{} \;\; (a,c)\in{}R $

#### **(3) 동치류(Equivalence Class)** : 동치 관계 $ R $에서 특정 원소 $ a $와 관계를 갖는 모든 원소들로 구성된 집합
#### $ \Rrightarrow{} [a]_{R} := \{x\in{}A \mid{} (a,x)\in{}R\} $

##### **(`PLUS`)** 하나의 동치 관계에서 생성되는 서로 다른 동치류들은 원래 집합을 서로 겹치지 않는 부분집합들로 분할함

<b></b>

> ## **함수(Function)**

#### **(1) 정의** : 집합 $ A $의 각 원소가 집합 $ B $의 원소 정확히 하나와 대응되는 이항 관계
#### $ \Rrightarrow{} f\subseteq{}A\times{}B $
#### $ \hspace{0.45cm} (\forall{}a\in{}A)(\exists{}!b\in{}B)\;\; (a,b)\in{}f $

##### **(`PLUS`)** 함수 $ f $에서 순서쌍 $ (a,b) $가 관계 $ f $에 포함되는 것은 함수값이 $ f(a)=b $임을 의미함
##### $ \hspace{0.15cm} \Rrightarrow{} f(a)=b \;\; \Leftrightarrow{} \;\; (a,b)\in{}f $

#### **(2) 구성 요소** :
##### $ \hspace{0.15cm} $ <img src="../img/01.2. Function (1).png" width="40%" height="40%"></img>
##### $ \hspace{0.15cm} $ ① 정의역(Domain) : 함수의 입력 원소들로 구성된 집합 $ A $
##### $ \hspace{0.15cm} $ ② 공역(Codomain) : 함수의 출력이 속하도록 지정된 집합 $ B $
##### $ \hspace{0.15cm} $ ③ 상(Image) : 정의역의 원소들이 함수를 통해 실제로 대응되는 함수값들의 집합
##### $ \hspace{0.45cm} f(A) := \{f(a)\mid{}a\in{}A\}\subseteq{}B $

<b></b>

> ## **함수의 분류(Classificaiton of Functions)**

#### **(1) 단사(Injective Function)** : 서로 다른 정의역의 원소가 서로 다른 함수값에 대응되는 함수
#### $ \Rrightarrow{} f(a_{1})=f(a_{2}) \;\; \Rightarrow{} \;\; a_{1}=a_{2} $
#### $ \hspace{0.45cm} \text{where } \, a_{1},a_{2}\in{}A $

#### **(2) 전사(Surjective Function)** : 함수의 상이 공역 전체와 동일한 함수
#### $ \Rrightarrow{} f(A)=B $
#### $ \Leftrightarrow{} (\forall{}b\in{}B)(\exists{}a\in{}A)\;\; f(a)=b $

#### **(3) 전단사(Bijective Function)** : 단사와 전사의 조건을 동시에 만족하는 함수
#### $ \Rrightarrow{} f \text{ is injective and surjective} $

<b></b>

> ## **상(Image)**

#### **(1) 정의** : 정의역의 부분집합 $ C $에 속하는 원소들의 함수값으로 구성된 집합
#### $ \Rrightarrow{} f(C) := \{f(x)\mid{}x\in{}C\} $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B,\;\; C\subseteq{}A $

#### **(2) 성질** :
##### $ \hspace{0.15cm} $ ① 공집합의 상은 공집합임
##### $ \hspace{0.45cm} f(\emptyset{})=\emptyset{} $
##### $ \hspace{0.15cm} $ ② 부분집합의 포함 관계는 상에서도 보존됨
##### $ \hspace{0.45cm} C\subseteq{}D \;\; \Rightarrow{} \;\; f(C)\subseteq{}f(D) $
##### $ \hspace{0.15cm} $ ③ 합집합의 상은 각 집합의 상의 합집합과 동일함
##### $ \hspace{0.45cm} f(C\cup{}D)=f(C)\cup{}f(D) $
##### $ \hspace{0.15cm} $ ④ 교집합의 상은 각 집합의 상의 교집합에 포함됨
##### $ \hspace{0.45cm} f(C\cap{}D)\subseteq{}f(C)\cap{}f(D) $

##### **(`WHY?`)** 서로 다른 입력이 동일한 함수값에 대응될 수 있으므로 $ f(C) $와 $ f(D) $에 공통으로 포함된 함수값이 반드시 $ C\cap{}D $의 원소에서 생성된 것은 아님

##### **(`PLUS`)** 함수 $ f $가 단사이면 교집합에 대해서도 등호가 성립함
##### $ \hspace{0.15cm} \Rrightarrow{} f(C\cap{}D)=f(C)\cap{}f(D) $

<b></b>

> ## **역상(Preimage;Inverse image)**

#### **(1) 정의** : 공역의 부분집합 $ D $에 포함되는 함수값을 갖는 정의역 원소들로 구성된 집합
#### $ \Rrightarrow{} f^{-1}(D) := \{a\in{}A\mid{}f(a)\in{}D\} $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B,\;\; D\subseteq{}B $
##### $ \hspace{0.15cm} $ <img src="../img/01.2. Function (5).png" width="40%" height="40%"></img>

#### **(2) 성질** :
##### $ \hspace{0.15cm} $ ① 공집합과 전체 공역의 역상은 각각 공집합과 전체 정의역임
##### $ \hspace{0.45cm} f^{-1}(\emptyset{})=\emptyset{},\;\; f^{-1}(B)=A $
##### $ \hspace{0.15cm} $ ② 부분집합의 포함 관계는 역상에서도 보존됨
##### $ \hspace{0.45cm} C\subseteq{}D \;\; \Rightarrow{} \;\; f^{-1}(C)\subseteq{}f^{-1}(D) $
##### $ \hspace{0.45cm} \text{where } \, C,D\subseteq{}B $
##### $ \hspace{0.15cm} $ ③ 합집합과 교집합은 역상에 의해 그대로 보존됨
##### $ \hspace{0.45cm} f^{-1}(C\cup{}D)=f^{-1}(C)\cup{}f^{-1}(D) $
##### $ \hspace{0.45cm} f^{-1}(C\cap{}D)=f^{-1}(C)\cap{}f^{-1}(D) $
##### $ \hspace{0.15cm} $ ④ 차집합은 역상에 의해 그대로 보존됨
##### $ \hspace{0.45cm} f^{-1}(C\setminus{}D)=f^{-1}(C)\setminus{}f^{-1}(D) $
##### $ \hspace{0.15cm} $ ⑤ 여집합은 역상에 의해 그대로 보존됨
##### $ \hspace{0.45cm} f^{-1}(B\setminus{}C)=A\setminus{}f^{-1}(C) $

##### **(`PLUS`)** 역상 표기 $ f^{-1}(D) $는 역함수의 존재 여부와 관계없이 모든 함수에 대해 정의됨

#### **(3) 예시** :
##### $ \hspace{0.15cm} $ ① 함수 $ f:\{1,2,3\}\rightarrow{}\{a,b,c,d\} $가 $ f(1)=a,\;\; f(2)=c,\;\; f(3)=c $일 때
##### $ \hspace{0.45cm} f^{-1}(\{c\})=\{2,3\} $
##### $ \hspace{0.45cm} f^{-1}(\{a,c\})=\{1,2,3\} $
##### $ \hspace{0.15cm} $ ② 함수 $ f:\mathbb{R}\rightarrow{}\mathbb{R},\;\; f(x)=x^{2} $일 때
##### $ \hspace{0.45cm} f^{-1}([1,4])=[-2,-1]\cup{}[1,2] $

> ## (Theorem 1-5) **Relationships between Image and Preimage**

#### 상과 역상을 연속해서 적용하면 원래 집합과 다음의 포함 또는 동일 관계를 만족한다.

#### **(1)** 정의역의 부분집합에 상을 적용한 뒤 다시 역상을 적용하면 원래 집합을 포함한다.
#### $ \Rrightarrow{} C\subseteq{}f^{-1}(f(C)) \;\; \hspace{0.45cm} \text{where } \, C\subseteq{}A $

#### **(2)** 공역의 부분집합에 역상을 적용한 뒤 다시 상을 적용하면 해당 집합과 함수의 전체 상의 교집합과 같다.
#### $ \Rrightarrow{} f(f^{-1}(D))=D\cap{}f(A) \;\; \hspace{0.45cm} \text{where } \, D\subseteq{}B $

#### **(3)** 함수가 단사이면 정의역의 모든 부분집합에 대해 상을 적용한 뒤 역상을 적용하면 원래 집합과 같다.
#### $ \Rrightarrow{} f^{-1}(f(C))=C $

#### **(4)** 함수가 전사이면 공역의 모든 부분집합에 대해 역상을 적용한 뒤 상을 적용하면 원래 집합과 같다.
#### $ \Rrightarrow{} f(f^{-1}(D))=D $

<b></b>

> ## **함수의 제한(Function Restriction)**

#### **(1) 정의** : 기존 함수의 대응 관계를 유지하면서 정의역을 특정 부분집합으로 제한한 함수
#### $ \Rrightarrow{} f|_{C}:C\rightarrow{}B $
#### $ \hspace{0.45cm} f|_{C}(x):=f(x) $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B,\;\; C\subseteq{}A $

<b></b>

> ## **합성 함수(Composite Function)**

#### **(1) 정의** : 한 함수의 함수값을 다른 함수의 입력으로 순차적으로 적용하여 구성한 함수
#### $ \Rrightarrow{} g\circ{}f:A\rightarrow{}C $
#### $ \hspace{0.45cm} (g\circ{}f)(a):=g(f(a)) $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B,\;\; g:B\rightarrow{}C $

#### **(2) 성질** :
##### $ \hspace{0.15cm} $ ① 함수의 합성은 결합법칙을 만족함
##### $ \hspace{0.45cm} (h\circ{}g)\circ{}f = h\circ{}(g\circ{}f) $
##### $ \hspace{0.15cm} $ ② 합성 함수의 역상은 각 함수의 역상을 역순으로 적용한 것과 동일함
##### $ \hspace{0.45cm} (g\circ{}f)^{-1}(D)=f^{-1}(g^{-1}(D)) $

<b></b>

> ## **항등 함수(Identity Function)**

#### **(1) 정의** : 집합의 각 원소를 자기 자신에 대응시키는 함수
#### $ \Rrightarrow{} \operatorname{id}_{A}:A\rightarrow{}A $
#### $ \hspace{0.45cm} \operatorname{id}_{A}(a):=a $

#### **(2) 성질** : 항등 함수를 다른 함수와 합성하면 원래 함수가 유지됨
#### $ \Rrightarrow{} f\circ{}\operatorname{id}_{A}=f $
#### $ \hspace{0.45cm} \operatorname{id}_{B}\circ{}f=f $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B $

<b></b>

> ## **역함수(Inverse Function)**

#### **(1) 정의** : 전단사 함수의 입력과 출력을 서로 반대로 대응시키는 함수
#### $ \Rrightarrow{} f^{-1}:B\rightarrow{}A $
#### $ \hspace{0.45cm} f^{-1}\circ{}f=\operatorname{id}_{A} $
#### $ \hspace{0.45cm} f\circ{}f^{-1}=\operatorname{id}_{B} $
#### $ \hspace{0.45cm} \text{where } \, f:A\rightarrow{}B \, \text{ is bijective} $

##### **(`PLUS`)** 역상(Preimage)은 집합에 적용되는 연산이고 역함수(Inverse Function)는 원소에 적용되는 함수이므로 서로 다른 개념임
##### $ \hspace{0.15cm} $ ① 역상
##### $ \hspace{0.45cm} f^{-1}(D),\;\; D\subseteq{}B $
##### $ \hspace{0.15cm} $ ② 역함수
##### $ \hspace{0.45cm} f^{-1}(b),\;\; b\in{}B $

#### **(2) 성질** :
##### $ \hspace{0.15cm} $ ① 역함수의 역함수는 원래 함수와 동일함
##### $ \hspace{0.45cm} (f^{-1})^{-1}=f $
##### $ \hspace{0.15cm} $ ② 두 전단사 함수의 합성에 대한 역함수는 각 역함수를 역순으로 합성한 것과 동일함
##### $ \hspace{0.45cm} (g\circ{}f)^{-1}=f^{-1}\circ{}g^{-1} $

#### **(3) 예시** :
##### $ \hspace{0.15cm} $ ① $ f:\mathbb{R}\rightarrow{}\mathbb{R},\;\; f(x)=3x+2 $일 때
##### $ \hspace{0.45cm} y=3x+2 \;\; \Leftrightarrow{} \;\; x=\frac{y-2}{3} $
##### $ \hspace{0.45cm} \therefore{} f^{-1}(y)=\frac{y-2}{3} $
##### $ \hspace{0.15cm} $ ② $ f:(0,\infty{})\rightarrow{}\mathbb{R},\;\; f(x)=\ln{}x $일 때
##### $ \hspace{0.45cm} y=\ln{}x \;\; \Leftrightarrow{} \;\; x=e^{y} $
##### $ \hspace{0.45cm} \therefore{} f^{-1}(y)=e^{y} $
