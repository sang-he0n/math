# CH00.1. **Measurable Structure**

> ## **측도론(Measure Theory)**

#### **(1) 정의** : 어떠한 집합의 부분집합 중 측정 대상으로 취급할 집합을 정하고, 해당 집합에 길이$ \cdot{} $넓이 등의 크기를 부여하는 방법을 다루는 이론
#### $ = $ 어떠한 집합을 측정할 수 있는지에 대한 구조와 해당 집합의 크기를 어떻게 측정할지를 일반적으로 다루는 이론

##### **(`PLUS`)** 측도론에서는 크게 어떠한 집합을 측정할 수 있는지를 나타내는 가측 구조와 해당 집합의 크기를 어떻게 측정할지를 나타내는 측도를 구분함

<b></b>

> ## **시그마-대수($ \sigma{} $-Algebra)**

#### **(1) 정의** : 어떠한 집합 $ U $의 부분집합 중 측정 대상으로 취급할 집합들을 모아놓은 집합족
#### $ \Rrightarrow{} \mathcal{F} \subseteq{} 2^{U} $
#### $ \hspace{0.45cm} \text{where } \, 2^{U} := \{A : A \subseteq{} U\} $

#### **(2) 공리** : 집합 $ U $ 위의 집합족 $ \mathcal{F} $가 시그마-대수일 경우 아래의 조건을 만족함
##### $ \hspace{0.15cm} $ ① 전체 집합 포함
##### $ \hspace{0.45cm} U \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ② 여집합에 대해 닫혀 있음
##### $ \hspace{0.45cm} A \in{} \mathcal{F} \; \Rightarrow{} \; A^{c} := U \setminus{} A \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ③ 가산 합집합에 대해 닫혀 있음
##### $ \hspace{0.45cm} A_{i} \in{} \mathcal{F} \; (\forall{} i \in{} \mathbb{N}) \; \Rightarrow{} \; \displaystyle{} \bigcup_{i=1}^{\infty{}} A_{i} \in{} \mathcal{F} $

#### **(3) 주요 성질**
##### $ \hspace{0.15cm} $ ① 공집합 포함
##### $ \hspace{0.45cm} \emptyset{} \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ② 가산 교집합에 대해 닫혀 있음
##### $ \hspace{0.45cm} A_{i} \in{} \mathcal{F} \; (\forall{} i \in{} \mathbb{N}) \; \Rightarrow{} \; \displaystyle{} \bigcap_{i=1}^{\infty{}} A_{i} \in{} \mathcal{F} $

#### **(4) 예시** :
##### $ \hspace{0.15cm} $ ① 자명한 시그마-대수(Trivial $ \sigma{} $-Algebra)
##### $ \hspace{0.45cm} \mathcal{F} = \{\emptyset{},U\} $
##### $ \hspace{0.15cm} $ ② 멱집합 시그마-대수(Power Set $ \sigma{} $-Algebra)
##### $ \hspace{0.45cm} \mathcal{F} = 2^{U} $

<b></b>

> ## **가측 공간(Measurable Space)**

#### **(1) 정의** : 어떠한 집합 $ U $와 해당 집합 위에 정의된 시그마-대수 $ \mathcal{F} $의 순서쌍
#### $ = $ 어떠한 부분집합을 측정 대상으로 취급할지를 정의한 구조
#### $ \Rrightarrow{} (U,\mathcal{F}) $

<b></b>

> ## **가측 집합(Measurable Set)**

#### **(1) 정의** : 가측 공간 $ (U,\mathcal{F}) $에서 시그마-대수 $ \mathcal{F} $에 포함되는 집합
#### $ = $ 해당 가측 공간에서 측정 대상으로 취급할 수 있는 집합
#### $ \Rrightarrow{} A \in{} \mathcal{F} $

<b></b>

> ## **생성 시그마-대수(Generated $ \sigma{} $-Algebra)**

#### **(1) 정의** : 어떠한 집합족 $ \mathcal{C} \subseteq{} 2^{U} $를 포함하는 시그마-대수 중 포함 관계에 대해 가장 작은 시그마-대수
#### $ \Rrightarrow{} \displaystyle{} \sigma{}(\mathcal{C}) := \bigcap{} \{ \mathcal{A} \subseteq{} 2^{U} \; : \; \mathcal{A} \, \text{ is a } \sigma{}\text{-algebra on } U \text{ and } \mathcal{C} \subseteq{} \mathcal{A} \}$
#### $ \hspace{0.45cm} \text{where } \, \mathcal{C} \subseteq{} 2^{U} $

##### **(`PLUS`) 생성족(Generating Family)** : 생성 시그마-대수 $ \sigma{}(\mathcal{C}) $를 생성하는 집합족 $ \, \mathcal{C} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ U=\{1,2,3\}, \;\; \mathcal{C}=\{\{1\}\} $가 존재할 때 $ \sigma{}(\mathcal{C}) $를 구하면
##### $ \hspace{0.45cm} \cdot{} \, \{1\} \in{} \sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1\}^{c} = \{2,3\} \in{} \sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \; \emptyset{} \in{} \sigma{}(\mathcal{C}), \;\; U\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \therefore{} \sigma{}(\mathcal{C})=\{\emptyset{},\{1\},\{2,3\},U\} $
##### $ \hspace{0.15cm} $ ② $ U=\{1,2,3,4\}, \;\; \mathcal{C}=\{\{1,2\},\{3\}\} $가 존재할 때 $ \sigma{}(\mathcal{C}) $를 구하면
##### $ \hspace{0.45cm} \cdot{} \, \{1,2\},\{3\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1,2\}\cup{}\{3\}=\{1,2,3\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1,2,3\}^{c}=\{4\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \therefore{} \sigma{}(\mathcal{C})=\{\emptyset{},\{1,2\},\{3\},\{4\},\{1,2,3\},\{1,2,4\},\{3,4\},U\} $

<b></b>

> ## **보렐 시그마-대수(Borel $ \sigma{} $-Algebra)**

#### **(1) 정의** : 실수 공간 $ \mathbb{R} $의 집합족 $ \{(a,b) : a,b \in{} \mathbb{R},\; a < b\} $에 의해 생성되는 시그마-대수
#### $ \Rrightarrow{} \mathcal{B}(\mathbb{R}) := \sigma{}(\{(a,b) : a,b \in{} \mathbb{R},\; a < b\}) $

##### **(`PLUS`)** 실수 공간에서는 서로 다른 형태의 구간 집합족을 사용하더라도 동일한 보렐 시그마-대수를 생성할 수 있음
##### $ \hspace{0.15cm} \Rrightarrow{} \mathcal{B}(\mathbb{R}) = \sigma{}(\{(-\infty{},a] : a \in{} \mathbb{R}\}) $

<b></b>

> ## **보렐 집합(Borel Set)**

#### **(1) 정의** : 실수 공간의 보렐 시그마-대수 $ \mathcal{B}(\mathbb{R}) $에 포함되는 집합
#### $ \Rrightarrow{} B \in{} \mathcal{B}(\mathbb{R}) $

<b></b>

> ## **(Theorem 0-1) Basic Borel Sets**

#### 실수 공간의 닫힌 구간, 한 점 집합, 반무한 구간 등은 보렐 집합이다.

#### $ \Rrightarrow{} [a,b], \; \{a\}, \; (-\infty{},a], \; [a,\infty{}) \in{} \mathcal{B}(\mathbb{R}) \;\; \text{where } \, a,b \in{} \mathbb{R}, \; a < b $

<b></b>

> ## **곱 시그마-대수(Product $ \sigma{} $-Algebra)**

#### **(1) 정의** : 두 가측 공간 $ (U,\mathcal{F}) $, $ (W,\mathcal{G}) $에 대해 곱집합 $ U \times{} W $ 위에서 $ A \times{} B $ 형태의 집합들에 의해 생성되는 시그마-대수
#### $ \Rrightarrow{} \mathcal{F} \otimes{} \mathcal{G} := \sigma{}(\{A \times{} B : A \in{} \mathcal{F},\; B \in{} \mathcal{G}\}) $

##### **(`PLUS`)** 곱 시그마-대수의 생성족은 $ A \times{} B $ 형태의 집합들로 구성되지만, 곱 시그마-대수의 모든 원소가 반드시 이러한 형태인 것은 아님
