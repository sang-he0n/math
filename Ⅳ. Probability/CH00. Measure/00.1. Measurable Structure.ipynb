# CH00.1. **Measurable Structure**

> ## **측도론(Measure Theory)**

#### **(1) 정의** : 집합의 부분 집합에 길이, 넓이, 부피, 확률 등의 크기 개념을 적용 및 일반화하는 이론

##### **(`PLUS`)** 확률론에서는 표본 공간의 부분 집합 중 확률을 부여할 대상을 정하고, 그 대상들 위에서 확률을 정의함

<b></b>

> ## **시그마-대수($\sigma{}$-Algebra)**

#### **(1) 정의** : 어떠한 집합 $ U $의 부분 집합들로 이루어진 집합족 중, 여집합 및 가산 합집합 연산에 닫혀 있고 공집합을 포함하는 집합족 $ \mathcal{F} $
#### $ \Rrightarrow{} \mathcal{F} \subseteq{} 2^{U} $

#### **(2) 공리** : 어떠한 집합 $ U $에 대응되는 시그마-대수 $ \mathcal{F} $가 존재할 때 아래를 만족해야함
##### $ \hspace{0.15cm} $ ① $ \emptyset{} \in{} \mathcal{F} $ 
##### $ \hspace{0.15cm} $ ② $ A \in{} \mathcal{F} \; \Rightarrow{} \; A^{c} := U \setminus{} A \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ③ $ \displaystyle{} A_{i} \in{} \mathcal{F} \; (\forall{}\,i) \; \Rightarrow{} \; \bigcup^{\infty{}}_{i=1} A_{i} \in{} \mathcal{F} $

#### **(3) 성질** : 어떠한 집합 $ U $에 대응되는 시그마-대수 $ \mathcal{F} $가 존재할 때 아래를 만족함
##### $ \hspace{0.15cm} $ ① $ U \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ② $ \displaystyle{} A_{i} \in{} \mathcal{F} \; \Rightarrow{} \; \bigcap^{\infty{}}_{i=1} A_{i} \in{} \mathcal{F} \;\; \text{ where } \, A_{i} \subset{} U $
##### $ \hspace{0.15cm} $ ③ $ A, B \in{} \mathcal{F} \; \Rightarrow{} \; (A \cup{} B) \in{} \mathcal{F}, \;\; (A \cap{} B) \in{} \mathcal{F}, \;\;  (A \cap{} B^{c}) \in{} \mathcal{F} $ 

<b></b>

> ## **생성 시그마-대수(Generated $ \sigma{} $-Algebra)**

#### **(1) 정의** : 집합 $ U $의 어떠한 집합족 $ \mathcal{C} $를 반드시 포함하는 시그마-대수 $ \mathcal{A} $ 중 가장 작은 포함 관계를 갖는 집합족
#### $ \Rrightarrow{} \displaystyle{} \sigma{}(\mathcal{C}) := \bigcap{} \{ \mathcal{A} \subseteq{} 2^{U} \; : \; \mathcal{A} \, \text{ is a } \sigma{}\text{-algebra on } U \text{ and } \mathcal{C} \subseteq{} \mathcal{A} \}$
#### $ \hspace{0.45cm} \text{where } \, \mathcal{C} \subseteq{} 2^{U} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ U=\{1,2,3\}, \;\; \mathcal{C}=\{\{1\}\} $가 존재할 때 $ \sigma{}(\mathcal{C}) $를 구하면
##### $ \hspace{0.45cm} \cdot{} \, \{1\} \in{} \sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1\}^{c} = \{2,3\} \in{} \sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \emptyset{} \in{} \sigma{}(\mathcal{C}), \;\; U\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \therefore{} \sigma{}(\mathcal{C})=\{\emptyset,\{1\},\{2,3\},U\} $
##### $ \hspace{0.15cm} $ ② $ U=\{1,2,3,4\}, \;\; \mathcal{C}=\{\{1,2\},\{3\}\} $가 존재할 때 $ \sigma{}(\mathcal{C}) $를 구하면
##### $ \hspace{0.45cm} \cdot{} \, \{1,2\},\{3\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1,2\}\cup{}\{3\}=\{1,2,3\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \cdot{} \, \{1,2,3\}^{c}=\{4\}\in{}\sigma{}(\mathcal{C}) $
##### $ \hspace{0.45cm} \therefore{} \sigma{}(\mathcal{C})=\{\emptyset{},\{1,2\},\{3\},\{4\},\{1,2,3\},\{1,2,4\},\{3,4\},U\} $

> ## **열린 집합(Open Set)**

#### **(1) 정의** : $ d $ 차원 실수 공간의 부분 집합 $ G $의 모든 원소(점) $ x $에 대해, $ x $ 주변의 충분히 작은 영역 $ B_{\epsilon{}}(x) $이 다시 $ G $ 안에 포함되는 집합
#### $ = $ 집합 내부의 임의의 점을 조금 움직여도 여전히 그 집합 안에 남아있는 집합
#### $ \Rrightarrow{} G \, \text{ is open } \;\; \Leftrightarrow{} \;\; \forall{}x\in{}G, \;\; \exists{}\epsilon{}>0 \;\; \text{ s.t. } \;\; B_{\epsilon{}}(x)\subseteq{}G $
#### $ \hspace{0.45cm} \text{where } \, B_{\epsilon{}}(x) := \{ y\in{}\mathbb{R}^{d} : \Vert{} y-x \Vert{} < \epsilon{} \} $

##### **(`PLUS`)** $ 1 $차원 실수 공간($ \mathbb{R} $)에서 열린 집합은 열린 구간들의 합집합으로 생각할 수 있음

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ G=(a,b)\subseteq{}\mathbb{R} $가 존재할 때 $ G $의 열린 집합 여부를 확인하면
##### $ \hspace{0.45cm} \text{let } \, \forall{}x\in{}G=(a,b), \;\; \epsilon{}:=\frac{\min{}\{x-a,b-x\}}{2}>0, \;\; \forall{}y\in{}B_{\epsilon{}}(x) $
##### $ \hspace{0.45cm} \cdot{} \, y\in{}B_{\epsilon{}}(x) \;\; \Leftrightarrow{} \;\; \Vert{}y-x\Vert{}<\epsilon{} \;\; \Leftrightarrow{} \;\; x-\epsilon{}<y<x+\epsilon{} $
##### $ \hspace{0.45cm} \cdot{} \, x-\epsilon{}=x-\frac{\min{}\{x-a,b-x\}}{2}>a $ 
##### $ \hspace{0.6cm} x+\epsilon{}=x+\frac{\min{}\{x-a,b-x\}}{2}<b $
##### $ \hspace{0.45cm} a<x-\epsilon{}<y<x+\epsilon{}<b \;\; \rightarrow{} \;\; y \in{}G \;\; \rightarrow{} \;\; B_{\epsilon{}}(x)\subseteq{}G $
##### $ \hspace{0.45cm} \therefore{} (a,b) \, \text{ is open} $
##### $ \hspace{0.15cm} $ ② $ G=[a,b]\subseteq{}\mathbb{R} $가 존재할 때 $ G $의 열린 집합 여부를 확인하면
##### $ \hspace{0.45cm} \text{let } \, x:=a\in{}G=[a,b], \;\; \forall{}\epsilon{}>0, \;\; y:=x-\frac{\epsilon{}}{2} $
##### $ \hspace{0.45cm} \cdot{} \, \Vert{}y-x\Vert{}=\Vert{}x-\frac{\epsilon{}}{2}-x\Vert{}=\frac{\epsilon{}}{2}<\epsilon{} \;\; \rightarrow{} \;\; y\in{}B_{\epsilon{}}(x) $
##### $ \hspace{0.45cm} \cdot{} \, y=x-\frac{\epsilon{}}{2}=a-\frac{\epsilon{}}{2}<a $
##### $ \hspace{0.45cm} y < a \;\; \Rightarrow{} \;\; y\notin{} G \;\; \rightarrow{} \;\; B_{\epsilon{}}(x)\not\subseteq{}G \;\; (\because{} \, B_{\epsilon{}}(x)\setminus{}G\neq{}\emptyset{}) $
##### $ \hspace{0.45cm} \therefore{} [a,b] \, \text{ is not open} $

> ## **보렐 시그마-대수(Borel $ \sigma{} $-Algebra)**

#### **(1) 정의** : $ d $ 차원 실수 공간의 열린 집합족 $ \mathcal{O}(\mathbb{R}^{d}) $로 생성된 시그마-대수
#### $ \approx{} $ 구간(interval)들로 생성한 시그마-대수
#### $ \Rrightarrow{} \mathcal{B}(\mathbb{R}^{d}) := \sigma{}\left(\mathcal{O}(\mathbb{R}^{d})\right) \subseteq{} 2^{\mathbb{R}^{d}} $
#### $ \hspace{0.45cm} \text{where } \, \mathcal{O}(\mathbb{R}^{d}) := \{ G\subseteq{}\mathbb{R}^{d} : G \, \text{ is open} \} $

#### **(2) 예시** :
##### $ \hspace{0.15cm} $ ① $ B=(a,b)\subseteq{}\mathbb{R} $가 존재할 때 $ B $의 보렐 집합 여부를 확인하면
##### $ \hspace{0.45cm} (a,b)\in{}\mathcal{O}(\mathbb{R}), \;\; \mathcal{O}(\mathbb{R})\subseteq{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \therefore{} (a,b)\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.15cm} $ ② $ B=[a,b]\subseteq{}\mathbb{R} $가 존재할 때 $ B $의 보렐 집합 여부를 확인하면
##### $ \hspace{0.45cm} (-\infty{},a), \;\; (b,\infty{})\in{}\mathcal{O}(\mathbb{R})\subseteq{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} (-\infty{},a)^{c}, \;\; (b,\infty{})^{c}\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} [a,b]=(-\infty{},a)^{c}\cap{}(b,\infty{})^{c}\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \therefore{} [a,b]\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.15cm} $ ③ $ B=\{a\}\subseteq{}\mathbb{R} $가 존재할 때 $ B $의 보렐 집합 여부를 확인하면
##### $ \hspace{0.45cm} \{a\}=[a,a], \;\; [a,a]\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \therefore{} \{a\}\in{}\mathcal{B}(\mathbb{R}) $

##### **(`PLUS`)** 보렐 시그마-대수는 열린 집합들로 생성되지만, 보렐 집합이 모두 열린 집합인 것은 아님

> ## **가측 공간과 가측 집합(Measurable Space and Measurable Set)**

#### **(1) 가측 공간(Measurable Space)** : 어떠한 집합 $ U $과 그 집합 위의 시그마-대수 $ \mathcal{F} $로 구성된 객체
#### $ \Rrightarrow{} (U, \mathcal{F}) $
#### $ \hspace{0.45cm} \text{where } \, U : \text{ set}, \;\; \mathcal{F} : \sigma{}\text{-algebra on } U $

#### **(2) 가측 집합(Measurable Set)** : 가측 공간 $ (U,\mathcal{F}) $에서 시그마-대수 $ \mathcal{F} $의 원소로 포함된 $ U $의 부분집합 
#### $ \Rrightarrow{} A \subseteq{} U, \;\; A \in{} \mathcal{F} $

<b></b>

> ## **곱 시그마-대수(Product $ \sigma{} $-Algebra)**

#### **(1) 정의** : 두 가측 공간 $ (U,\mathcal{F}) $, $ (V,\mathcal{G}) $가 존재할 때, 가측 집합 $ A\times{}B $들이 생성하는 시그마-대수
#### $ \Rrightarrow{} \mathcal{F}\otimes{}\mathcal{G}:=\sigma{}(\{A\times{}B:A\in{}\mathcal{F},B\in{}\mathcal{G}\}) $
