# CH00.5. **Measure Transformation**

> ## **제한 측도(Restricted Measure)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 가측 집합 $ E \in{} \mathcal{F} $에 측도 $ \mu{} $를 제한하여 정의한 측도
#### $ = $ 전체 공간의 측도 중 집합 $ E $에 포함되는 부분만을 측정하는 측도
#### $ \Rrightarrow{} \mu{}|_{E}(A) := \mu{}(A \cap{} E) $
#### $ \hspace{0.45cm} \text{where } \, A \in{} \mathcal{F},\;\; E \in{} \mathcal{F} $

##### **(`PLUS`)** 제한된 공간 $ E $ 자체를 새로운 전체 공간으로 사용할 경우 제한 시그마-대수는 아래와 같이 정의할 수 있음
##### $ \hspace{0.15cm} \mathcal{F}|_{E} := \{A \cap{} E : A \in{} \mathcal{F}\} $
##### $ \hspace{0.15cm} \Rrightarrow{} (E,\mathcal{F}|_{E},\mu{}|_{E}) $

<b></b>

> ## **푸시포워드 측도(Pushforward Measure; Image Measure)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $과 가측 공간 $ (W,\mathcal{G}) $ 사이의 가측 사상 $ f $를 통해 정의역의 측도 $ \mu{} $를 공역 $ W $ 위의 측도로 이동시킨 측도
#### $ = $ 공역의 가측 집합 $ B $에 대해 그 역상 $ f^{-1}(B) $의 기존 측도를 값으로 부여하는 측도
#### $ \Rrightarrow{} f_{\#}\mu{}(B) := \mu{}(f^{-1}(B)) $
#### $ \hspace{0.45cm} \text{where } \, f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}) \, \text{ is measurable},\;\; B \in{} \mathcal{G} $
#### $ \Rrightarrow{} f_{\#}\mu{} : \mathcal{G} \rightarrow{} [0,\infty{}] $

<b></b>

> ## **곱 측도(Product Measure)**

#### **(1) 정의** : 두 측도 공간 $ (U,\mathcal{F},\mu{}) $, $ (W,\mathcal{G},\nu{}) $를 결합한 곱 공간 $ U \times{} W $ 위에서 정의되는 측도
#### $ \Rrightarrow{} \mu{} \otimes{} \nu{} : \mathcal{F} \otimes{} \mathcal{G} \rightarrow{} [0,\infty{}] $

#### **(2) 직사각형 집합에 대한 측도** : 가측 집합 $ A \in{} \mathcal{F} $, $ B \in{} \mathcal{G} $의 곱집합에 각 측도의 곱을 부여함
#### $ \Rrightarrow{} (\mu{} \otimes{} \nu{})(A \times{} B) = \mu{}(A)\nu{}(B) $

##### **(`PLUS`)** 두 측도가 시그마-유한 측도인 경우 위 성질을 만족하는 곱 측도가 곱 시그마-대수 $ \mathcal{F}\otimes{}\mathcal{G} $ 위에서 유일하게 결정됨

<b></b>

> ## **주변 측도(Marginal Measure)**

#### **(1) 정의** : 곱 가측 공간 $ (U \times{} W,\mathcal{F}\otimes{}\mathcal{G}) $ 위의 측도 $ \rho{} $에서 특정 좌표만을 남겨 정의한 측도
#### $ = $ 곱 공간의 측도 $ \rho{} $를 좌표 사상(Projection Mapping)을 통해 각 좌표 공간으로 푸시포워드한 측도
##### $ \hspace{0.15cm} $ ① 첫 번째 좌표 $ U $에 대한 주변 측도
##### $ \hspace{0.45cm} \rho_{U} := (\pi_{U})_{\#}\rho{},\;\; \rho_{U}(A)=\rho{}(A \times{} W) $
##### $ \hspace{0.45cm} \text{where } \, \pi_{U}(u,w):=u,\;\; A \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ② 두 번째 좌표 $ W $에 대한 주변 측도
##### $ \hspace{0.45cm} \rho_{W} := (\pi_{W})_{\#}\rho{},\;\; \rho_{W}(B)=\rho{}(U \times{} B) $
##### $ \hspace{0.45cm} \text{where } \, \pi_{W}(u,w):=w,\;\; B \in{} \mathcal{G} $

##### **(`PLUS`)** 주변 측도는 푸시포워드 측도의 특수한 경우임

<b></b>

> ## **측도 간 관계(Relation between Measures)**

#### **(1) 주요 관계** :
##### $ \hspace{0.15cm} $ ① 절대연속성(Absolute Continuity) : 측도 $ \mu{} $가 $ 0 $을 부여하는 모든 가측 집합에 측도 $ \nu{} $ 역시 $ 0 $을 부여하는 관계
##### $ \hspace{0.45cm} \nu{} \ll{} \mu{} $
##### $ \hspace{0.45cm} \Leftrightarrow{} \mu{}(A)=0 \;\; \Rightarrow{} \;\; \nu{}(A)=0 \;\;\; \forall{} A \in{} \mathcal{F} $
##### $ \hspace{0.15cm} $ ② 특이성(Singularity) : 두 측도 $ \mu{} $, $ \nu{} $가 서로 겹치지 않는 가측 영역에 각각 집중되어 있는 관계
##### $ \hspace{0.45cm} \mu{} \perp{} \nu{} $
##### $ \hspace{0.45cm} \Leftrightarrow{} \exists{} E \in{} \mathcal{F} \;\; \text{s.t. } \, \mu{}(E^{c})=0,\;\; \nu{}(E)=0 $
