# CH00.3. **Measure Space**

> ## **측도 공간(Measure Space)**

#### **(1) 정의** : 가측 공간 $ (U,\mathcal{F}) $과 그 위에 정의된 측도 $ \mu{} $로 구성된 객체
#### $ \Rrightarrow{} (U,\mathcal{F},\mu{}) $
#### $ \hspace{0.45cm} \text{where } \, \mathcal{F} : \sigma{}\text{-algebra on } U,\;\; \mu{} : \mathcal{F} \rightarrow{} [0,\infty{}] $

<b></b>

> ## **영측도 집합(Null Set)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 측도가 $ 0 $인 가측 집합
#### $ \Rrightarrow{} A \in{} \mathcal{F},\;\; \mu{}(A)=0 $

##### **(`PLUS`)** 영측도 집합은 측도 $ \mu{} $에 대해 정의되므로 동일한 집합이라도 어떠한 측도를 사용하는지에 따라 영측도 집합 여부가 달라질 수 있음

<b></b>

> ## **거의 어디서나(Almost Everywhere; a.e.)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 어떠한 성질 $ p(u) $가 측도 $ 0 $인 가측 집합을 제외한 모든 원소에서 성립하는 상태
#### $ \Rrightarrow{} p(u) \text{ holds } \mu{}\text{-a.e.} $
#### $ \Leftrightarrow{} \exists{} N \in{} \mathcal{F} \;\; \text{s.t. } \, \mu{}(N)=0,\;\; p(u) \text{ holds for all } u \in{} U \setminus{} N $

##### **(`PLUS`)** 성질 $ p(u) $가 성립하지 않는 예외 집합 $ E_{p} $가 가측 집합일 경우 아래와 같이 표현할 수 있음
##### $ \hspace{0.15cm} E_{p} := \{u \in{} U : p(u) \text{ does not hold}\} $
##### $ \hspace{0.15cm} \Rrightarrow{} p(u) \text{ holds } \mu{}\text{-a.e.} \Leftrightarrow{} \mu{}(E_{p})=0 $

<b></b>

> ## **완비 측도 공간(Complete Measure Space)**

#### **(1) 정의** : 모든 영측도 집합의 모든 부분집합까지 가측 집합으로 포함하는 측도 공간
#### $ = $ 어떠한 영측도 집합 $ N $의 부분집합 $ A $도 가측 집합으로 취급할 수 있는 측도 공간
#### $ \Rrightarrow{} (U,\mathcal{F},\mu{}) \text{ is complete} $
#### $ \Leftrightarrow{} N \in{} \mathcal{F},\;\; \mu{}(N)=0,\;\; A \subseteq{} N \; \Rightarrow{} \; A \in{} \mathcal{F} \;\; (\forall{} N,\;\forall{} A \subseteq{} N) $

##### **(`PLUS`)** 일반적인 측도 공간에서는 영측도 집합 $ N $이 가측하더라도 그 부분집합 $ A \subseteq{} N $이 반드시 가측 집합인 것은 아님

<b></b>

> ## **측도 공간의 완비화(Completion of Measure Space)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $를 기존 측도 구조를 보존하면서 완비 측도 공간으로 확장하는 과정
#### $ \Rrightarrow{} (U,\mathcal{F},\mu{}) \mapsto{} (U,\overline{\mathcal{F}},\overline{\mu{}}) $
#### $ \hspace{0.45cm} \text{where } \, \overline{\mathcal{F}} : \text{completed } \sigma{}\text{-algebra},\;\; \overline{\mu{}} : \text{completed measure} $

#### **(2) 완비화 과정** :
##### $ \hspace{0.15cm} $ ① 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 영측도 집합 $ Z \in{} \mathcal{F} $를 확인함
##### $ \hspace{0.45cm} \mu{}(Z)=0 $
##### $ \hspace{0.15cm} $ ② 영측도 집합 $ Z $의 임의의 부분집합 $ N \subseteq{} Z $를 고려함
##### $ \hspace{0.45cm} N \subseteq{} Z,\;\; \mu{}(Z)=0 $
##### $ \hspace{0.15cm} $ ③ 기존 가측 집합 $ A \in{} \mathcal{F} $에 영측도 집합의 부분집합 $ N $을 추가하여 새로운 집합 $ E $를 구성함
##### $ \hspace{0.45cm} E := A \cup{} N $
##### $ \hspace{0.15cm} $ ④ 이와 같이 구성되는 모든 집합 $ E $를 포함하도록 기존 시그마-대수 $ \mathcal{F} $를 완비화된 시그마-대수 $ \overline{\mathcal{F}} $로 확장함
##### $ \hspace{0.45cm} E \in{} \overline{\mathcal{F}},\;\; \mathcal{F} \subseteq{} \overline{\mathcal{F}} $
##### $ \hspace{0.15cm} $ ⑤ 기존 측도 $ \mu{} $를 완비화된 시그마-대수 $ \overline{\mathcal{F}} $ 위의 완비화된 측도 $ \overline{\mu{}} $로 확장함
##### $ \hspace{0.45cm} \mu{} \mapsto{} \overline{\mu{}} $
##### $ \hspace{0.15cm} $ ⑥ 완비화된 측도 공간에서는 모든 영측도 집합의 모든 부분집합이 가측 집합에 포함됨
##### $ \hspace{0.45cm} M \in{} \overline{\mathcal{F}},\;\; \overline{\mu{}}(M)=0,\;\; B \subseteq{} M \;\; \Rightarrow{} \;\; B \in{} \overline{\mathcal{F}} $
##### $ \hspace{0.45cm} \therefore{} (U,\overline{\mathcal{F}},\overline{\mu{}}) \;\; \text{is complete measure space} $

#### **(3) 완비화된 시그마-대수(Completed $ \sigma{} $-Algebra)** : 기존 가측 집합에 영측도 집합의 임의의 부분집합을 추가하여 구성한 시그마-대수
#### $ \Rrightarrow{} \overline{\mathcal{F}} := \{A \cup{} N : A \in{} \mathcal{F},\;\; N \subseteq{} Z,\;\; Z \in{} \mathcal{F},\;\; \mu{}(Z)=0\} $

#### **(4) 완비화된 측도(Completed Measure)** : 완비화된 시그마-대수 위에서 기존 측도 $ \mu{} $를 보존하도록 확장한 측도
#### $ \Rrightarrow{} \overline{\mu{}}(A\cup{}N):=\mu{}(A) $
#### $ \hspace{0.45cm} \text{where } \, A\in{}\mathcal{F},\;\; N\subseteq{}Z,\;\; Z\in{}\mathcal{F},\;\; \mu{}(Z)=0 $

<b></b>

> ## **르베그 측도(Lebesgue Measure)**

#### **(1) 정의** : 실수 공간에서 구간의 길이를 일반화하여 보렐 집합 및 그 영측도 부분집합까지 측정할 수 있도록 정의된 완비 측도
#### $ \Rrightarrow{} \lambda{} : \mathcal{L}(\mathbb{R}) \rightarrow{} [0,\infty{}] $
#### $ \hspace{0.45cm} \text{where } \, \mathcal{L}(\mathbb{R}) : \text{Lebesgue } \sigma{}\text{-algebra} $

#### **(2) 구간에 대한 측도**
#### $ \Rrightarrow{} \lambda{}([a,b]) = b-a \;\; \text{where } \, a,b \in{} \mathbb{R},\;\; a \leq{} b $

##### **(`PLUS`)** 르베그 시그마-대수 $ \mathcal{L}(\mathbb{R}) $는 보렐 시그마-대수 $ \mathcal{B}(\mathbb{R}) $를 영측도 집합의 부분집합까지 포함하도록 완비화하여 얻는 시그마-대수임
##### $ \hspace{0.15cm} \Rrightarrow{} \mathcal{B}(\mathbb{R}) \subseteq{} \mathcal{L}(\mathbb{R}) $
