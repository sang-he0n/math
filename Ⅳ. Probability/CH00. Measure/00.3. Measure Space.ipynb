# CH00.3. **Measure Space**

> ## **측도 공간(Measure space)**

#### **(1) 정의** : 어떠한 집합과 그 집합과 대응되는 시그마-대수, 측도와 함께 구성된 객체
#### $ \Rrightarrow{} (U, \mathcal{F}, \mu{}) $ 
#### $ \hspace{0.45cm} \text{where } \, U : \text{set}, \;\; \mathcal{F} : \sigma{}\text{-algebra on } U, \;\; \mu{}:\mathcal{F}\to{}[0,\infty{}] \;\; (\text{measure}) $

<b></b>

> ## **영측도 집합(Null Set)**

#### **(1) 정의** : 측도가 $ 0 $인 가측 집합
#### $ \Rrightarrow{} \mu{}(A)=0 \;\; \text{ where } \, A \in{} \mathcal{F} $

<b></b>

> ## **거의 어디서나(Almost Everywhere; a.e.)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 판정되는 성질 $ p(u) $가 영측도 집합을 제외하고 성립하는 상태
#### $ = U $의 원소 중 성질 $ \, p(u) $를 만족하지 않는 원소들의 집합(예외 집합)이 측도 $ 0 $인 상태
#### $ \Rrightarrow{} p(u) \, \text{ holds } \mu{}\text{-a.e.} \iff \mu{}(E_p)=0 $
#### $ \hspace{0.45cm} \text{where } \, E_p:=\{u\in{}U:p(u) \text{ does not hold}\} (\text{exception set}) $

<b></b>

> ## **완비 측도 공간(Complete Measure Space)**

#### **(1) 정의** : 모든 영측도 집합의 부분집합까지 가측 집합으로 포함하는 측도 공간
#### $ = $ 영측도 집합 $ N $ 안에 포함된 어떠한 부분집합 $ A $도 가측 집합으로 취급할 수 있는 측도 공간
#### $ \Rrightarrow{} (U,\mathcal{F},\mu{}) \, \text{ is complete } \iff N\in{}\mathcal{F}, \;\; \mu{}(N)=0, \;\; A\subseteq{}N \; \Rightarrow{} \; A\in{}\mathcal{F} $

<b></b>

> ## **측도 공간의 완비화(Completion of Measure Space)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $를 완비 측도 공간 $ (U,\overline{\mathcal{F}},\overline{\mu{}}) $으로 확장하는 과정
#### $ = $ 기존 시그마-대수 $ \mathcal{F} $에 영측도 집합 내부의 모든 부분집합을 추가하여 완비성을 만족하도록 만드는 과정
#### $ \Rrightarrow{} (U,\mathcal{F},\mu{}) \mapsto{} (U,\overline{\mathcal{F}},\overline{\mu{}}) $

#### **(2) 과정** :
##### $ \hspace{0.15cm} $ ① 기존 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 영측도 집합 $ N $을 정의
##### $ \hspace{0.45cm} N\in{}\mathcal{F}, \;\; \mu{}(N)=0 $
##### $ \hspace{0.15cm} $ ② 영측도 집합 $ N $의 모든 부분집합 $ A $를 확장된 시그마-대수 $ \overline{\mathcal{F}} $에 추가
##### $ \hspace{0.45cm} A\subseteq{}N \;\; \Rightarrow{} \;\; A\in{}\overline{\mathcal{F}} $
##### $ \hspace{0.15cm} $ ③ 기존 시그마-대수 $ \mathcal{F} $를 확장된 시그마-대수 $ \overline{\mathcal{F}} $ 안에 포함
##### $ \hspace{0.45cm} \mathcal{F}\subseteq{}\overline{\mathcal{F}} $
##### $ \hspace{0.15cm} $ ④ 기존 측도 $ \mu{} $를 확장된 시그마-대수 $ \overline{\mathcal{F}} $ 위의 측도 $ \overline{\mu{}} $로 확장
##### $ \hspace{0.45cm} \overline{\mu{}}|_{\mathcal{F}}=\mu{} $
##### $ \hspace{0.15cm} $ ⑤ 확장된 측도 공간 $ (U,\overline{\mathcal{F}},\overline{\mu{}}) $은 완비 측도 공간
##### $ \hspace{0.45cm} N\in{}\overline{\mathcal{F}}, \;\; \overline{\mu{}}(N)=0, \;\; A\subseteq{}N \;\; \Rightarrow{} \;\; A\in{}\overline{\mathcal{F}} $
##### $ \hspace{0.45cm} \therefore{} (U,\overline{\mathcal{F}},\overline{\mu{}}) \;\; \text{is complete measure space} $
