# CH00.2. **Measure**

> ## **측도(Measure)**

#### **(1) 정의** : 가측 공간 $ (U,\mathcal{F}) $에서 시그마-대수 $ \mathcal{F} $ 위에 정의된 확장 비음수 실수값 집합 함수(Set Function)
#### $ = $ 가측 집합에 크기, 길이, 질량, 확률 등의 값을 부여하는 함수
#### $ \Rrightarrow{} \mu{} : \mathcal{F} \rightarrow{} [0,\infty{}] $
#### $ \hspace{0.45cm} A \mapsto{} \mu{}(A) \;\; \text{where } \, A \in{} \mathcal{F} $

##### **(`PLUS`)** $ [0,\infty{}] := [0,\infty{}) \cup{} \{\infty{}\} \;\; $ (측도는 무한한 값을 가질 수 있음)

#### **(2) 공리** : 집합 함수 $ \mu{} : \mathcal{F} \rightarrow{} [0,\infty{}] $가 측도일 경우 아래의 조건을 만족함
##### $ \hspace{0.15cm} $ ① 공집합의 측도는 $ 0 $
##### $ \hspace{0.45cm} \mu{}(\emptyset{}) = 0 $
##### $ \hspace{0.15cm} $ ② 서로소인 가측 집합열에 대해 가산 가법성(Countable Additivity)을 만족
##### $ \hspace{0.45cm} A_{i} \cap{} A_{j} = \emptyset{} \;\; (\forall{} i,j \in{} \mathbb{N},\; i \neq{} j) \;\; \Rightarrow{} \;\; \displaystyle{} \mu{}\left(\bigcup_{i=1}^{\infty{}} A_{i}\right) = \sum_{i=1}^{\infty{}} \mu{}(A_{i}) \;\; \text{where } \, A_{i} \in{} \mathcal{F} $

<b></b>

> ## **(Theorem 0-2) Basic Properties of Measure**

#### **(1)** 포함관계에 따라 측도도 증가한다.
#### $ \Rrightarrow{} A \subseteq{} B,\;\; A,B \in{} \mathcal{F} \; \Rightarrow{} \; \mu{}(A) \leq{} \mu{}(B) $

#### **(2)** 서로소인 유한 개의 가측 집합의 합집합에 대한 측도는 각 측도의 합과 같다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcup_{i=1}^{n} A_{i}\right) = \sum_{i=1}^{n} \mu{}(A_{i}) $
#### $ \hspace{0.45cm} \text{where } \, A_{i} \in{} \mathcal{F},\;\; A_{i} \cap{} A_{j} = \emptyset{} \;\; (i \neq{} j) $

#### **(3)** 가측 집합들의 합집합에 대한 측도는 각 측도의 합보다 크지 않다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcup_{i=1}^{\infty{}} A_{i}\right) \leq{} \sum_{i=1}^{\infty{}} \mu{}(A_{i}) \;\; \text{where } \, A_{i} \in{} \mathcal{F} $

#### **(4)** 포함관계에 있는 두 가측 집합에서 작은 집합의 측도가 유한할 경우 차집합의 측도는 두 측도의 차와 같다.
#### $ \Rrightarrow{} A \subseteq{} B,\;\; \mu{}(A) < \infty{} \;\; \Rightarrow{} \;\; \mu{}(B \setminus{} A) = \mu{}(B) - \mu{}(A) $
#### $ \hspace{0.45cm} \text{where } \, A,B \in{} \mathcal{F} $

<b></b>

> ## **(Theorem 0-3) Continuity of Measure**

#### **(1)** 증가하는 가측 집합열의 합집합에 대한 측도는 각 집합의 측도의 극한과 같다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcup_{n=1}^{\infty{}} A_{n}\right) = \lim_{n \rightarrow{} \infty{}} \mu{}(A_{n}) $
#### $ \hspace{0.45cm} \text{s.t. } \, A_{1} \subseteq{} A_{2} \subseteq{} \cdots{},\;\; A_{n} \in{} \mathcal{F} $

#### **(2)** 감소하는 가측 집합열에서 첫 번째 집합의 측도가 유한할 경우 교집합에 대한 측도는 각 집합의 측도의 극한과 같다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcap_{n=1}^{\infty{}} A_{n}\right) = \lim_{n \rightarrow{} \infty{}} \mu{}(A_{n}) $
#### $ \hspace{0.45cm} \text{s.t. } \, A_{1} \supseteq{} A_{2} \supseteq{} \cdots{},\;\; A_{n} \in{} \mathcal{F},\;\; \mu{}(A_{1}) < \infty{} $

<b></b>

> ## **측도의 유한성(Finiteness of Measure)**

#### **(1) 유한 측도(Finite Measure)** : 전체 공간 $ U $의 측도가 유한한 측도
#### $ \Rrightarrow{} \mu{}(U) < \infty{} $

#### **(2) 시그마-유한 측도($ \sigma{} $-Finite Measure)** : 전체 공간 $ U $를 유한한 측도를 갖는 가측 집합들의 가산 합집합으로 표현할 수 있는 측도
#### $ \Rrightarrow{} \displaystyle{} U = \bigcup_{n=1}^{\infty{}} U_{n} $
#### $ \hspace{0.45cm} \text{where } \, U_{n} \in{} \mathcal{F},\;\; \mu{}(U_{n}) < \infty{} \;\; (\forall{} n \in{} \mathbb{N}) $

##### **(`PLUS`)** 모든 유한 측도는 시그마-유한 측도임

<b></b>

> ## **보렐 측도(Borel Measure)**

#### **(1) 정의** : 실수 공간의 보렐 시그마-대수 $ \mathcal{B}(\mathbb{R}) $ 위에 정의된 측도
#### $ \Rrightarrow{} \mu{} : \mathcal{B}(\mathbb{R}) \rightarrow{} [0,\infty{}] $

##### **(`PLUS`)** 보렐 측도는 측도의 종류라기보다 측도의 정의역이 보렐 시그마-대수인 경우를 의미함

<b></b>

> ## **계수 측도(Counting Measure)**

#### **(1) 정의** : 가측 집합 $ A $에 포함되는 원소의 개수를 값으로 부여하는 측도
#### $ \Rrightarrow{} \mu{}(A) := \begin{cases} |A| & \text{if } \, A \text{ is finite} \\ \infty{} & \text{if } \, A \text{ is infinite} \end{cases} $
#### $ \hspace{0.45cm} \text{where } \, A \in{} \mathcal{F} $
