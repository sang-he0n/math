# CH00.2. **Measure**

> ## **측도(Measure)**

#### **(1) 정의** : 가측 공간 $ (U,\mathcal{F}) $에서 시그마-대수 $ \mathcal{F} $ 위에 정의된 확장 비음수 실수값 집합 함수(set function)
#### $ = $ 가측 집합에 크기, 길이, 질량, 확률 등을 부여하는 함수
#### $ \Rrightarrow{} \mu{} : \mathcal{F} \rightarrow{} [0,\infty{}] \;\; \Leftrightarrow{} \;\; A \mapsto{} \mu{}(A) \;\; \text{ where } \, A\in{}\mathcal{F} $

#### **(2) 공리** : 어떠한 시그마-대수 $ \mathcal{F} $에 대응되는 측도 $ \mu{} $가 존재할 때 아래를 만족해야함
##### $ \hspace{0.15cm} $ ① $ \mu{}(A) \geq{} 0 $
##### $ \hspace{0.15cm} $ ② $ \mu{}(\empty{}) = 0 $ 
##### $ \hspace{0.15cm} $ ③ $ \displaystyle{} A_{i} \cap{} A_{j} = \empty{} \;\; (\forall{} \, i \neq{} j) \; \Rightarrow{} \; \mu{}(\bigcup^{\infty{}}_{i=1}A_{i}) = \sum^{\infty{}}_{i=1} \mu{}(A_{i}) \;\; \text{ where } \, A_{i} \in{} \mathcal{F} $ 

<b></b>

> ## (Theorem 0-1) **Basic Properties of Measure**

#### **(1)** 포함관계에 따라 측도도 증가한다.
#### $ \Rrightarrow{} A \subseteq{} B,\;\; A,B \in{}\mathcal{F} \; \Rightarrow{} \; \mu{}(A)\le{}\mu{}(B) $

#### **(2)** 서로소인 유한 개의 가측집합의 합집합에 대한 측도는 각 측도의 합과 같다. 
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcup_{i=1}^{n}A_i\right)=\sum_{i=1}^{n}\mu{}(A_i) \;\; \text{ where } \, A_{i}\cap{}A_{j}=\emptyset{} \;\; (i\neq{}j), \;\; i,j=1,\dots,n $

#### **(3)** 가측 집합들의 합집합에 대한 측도는 각 측도의 합보다 크지 않다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\left(\bigcup_{i=1}^{\infty{}} A_{i}\right)\le{}\sum_{i=1}^{\infty{}}\mu{}(A_{i}) \;\; \text{ where } \, A_{i}\in{}\mathcal{F} $

#### **(4)** 포함관계에 있는 두 가측 집합의 경우, 차집합 측도를 뺄셈으로 표현할 수 있다.
#### $ \Rrightarrow{} A \subseteq{} B,\;\; A,B \in{}\mathcal{F},\;\; \mu{}(A)<\infty{} \;\; \Rightarrow{} \;\; \mu{}(B\setminus{}A)=\mu{}(B)-\mu{}(A) $

<b></b>

> ## (Theorem 0-2) **Continuity of Measure**

#### **(1)** 어떠한 집합열 $ A_{1}\subseteq{}A_{2}\subseteq{}\cdots{} $이 존재할 때 그 합집합의 측도는 각 집합의 측도의 극한과 같다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\Big({}\bigcup_{n=1}^{\infty{}}A_{n}\Big) = \lim_{n\to{}\infty{}}\mu{}(A_{n}) \;\; \text{ s.t. } \, A_{1}\subseteq{}A_{2}\subseteq{}\cdots{}, \;\; A_{n}\in{}\mathcal{F} $

#### **(2)** 어떠한 집합열 $ A_{1}\supseteq{}A_{2}\supseteq{}\cdots{} $이 존재하고 $ \mu{}(A_{1})<\infty{} $일 때 그 교집합의 측도는 각 집합의 측도의 극한과 같다.
#### $ \Rrightarrow{} \displaystyle{} \mu{}\Big({}\bigcap_{n=1}^{\infty{}}A_{n}\Big) = \lim_{n\to{}\infty{}}\mu{}(A_{n}) \;\; \text{ s.t. } \, A_{1}\supseteq{}A_{2}\supseteq{}\cdots{}, \;\; A_{n}\in{}\mathcal{F}, \;\; \mu{}(A_{1})<\infty{} $

<b></b>

> ## **측도의 분류 : 크기(Measure Classification by Size)**

#### **(1) 유한 측도(Finite Measure)** : 전체 공간의 측도가 유한한 측도
#### $ \Rrightarrow{} \mu{}(U)<\infty{} \;\; \text{ where } \, \mu{}:\mathcal{F}\rightarrow{}[0,\infty{}] $

#### **(2) 시그마-유한 측도($\sigma{}$-Finite Measure)** : 전체 공간을 유한 측도를 갖는 가측 집합들의 가산합집합으로 덮을 수 있는 측도
#### $ \Rrightarrow{} \displaystyle{} U=\bigcup_{n=1}^{\infty{}}U_n, \;\; \mu{}(U_n)<\infty{}, \;\;\;\; U_n\in{}\mathcal{F} $

<b></b>

> ## **측도의 분류 : 정의역(Measure Classification by Domain)**

#### **(1) 보렐 측도(Borel Measure)** : $ d $ 차원 실수 공간의 보렐 시그마-대수 $ \mathcal{B}(\mathbb{R}^{d}) $ 위에 정의된 측도
#### $ \Rrightarrow{} \mu{}:\mathcal{B}(\mathbb{R}^{d})\rightarrow{}[0,\infty{}] \;\; \text{where } \, \mathcal{B}(\mathbb{R}^{d})=\sigma{}\left(\mathcal{O}(\mathbb{R}^{d})\right) $

<b></b>

> ## **측도의 분류 : 기준 측도(Reference Measure)**

#### **(1) 계수 측도(Counting Measure)** : 가측 집합 안에 원소의 개수를 정의하는 측도
#### $ \Rrightarrow{} \mu{}(A) = \begin{cases} |A| & \text{if } \, A \, \text{ is finite} \\ \infty{} & \text{if } \, A \, \text{ is infinite} \end{cases} \;\; \text{ where } \, A \in{} \mathcal{F} $

#### **(2) 르베그 측도(Lebesgue Measure)** : $ d $ 차원 실수 공간에서 길이, 넓이, 부피를 일반화한 측도
#### $ \Rrightarrow{} \lambda{}([a,b])=b-a $
