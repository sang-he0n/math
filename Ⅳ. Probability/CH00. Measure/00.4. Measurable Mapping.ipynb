# CH00.4. **Measurable Mapping**

> ## **가측 사상(Measurable Mapping)**

#### **(1) 정의** : 두 가측 공간 사이의 함수 $ f $에 대해, 공역의 모든 가측 집합 $ B $의 역상 $ f^{-1}(B) $가 정의역의 시그마-대수에 포함되는 사상(함수) $ \, f $
#### $ \Rrightarrow{} f : (U,\mathcal{F}) \rightarrow{} (W,\mathcal{G}) \, \text{ is measurable} \;\; \Leftrightarrow{} \;\; f^{-1}(B) \in{} \mathcal{F} \;\;\;\; \forall{} B \in{} \mathcal{G} $
#### $ \hspace{0.45cm} \text{where } \, \mathcal{F} \, \text{ is a } \sigma{}\text{-algebra on } U, \;\; \mathcal{G} \, \text{ is a } \sigma{}\text{-algebra on } W $

<b></b>

> ## **실수값 가측 함수(Real-Valued Measurable Function)**

#### **(1) 정의** : 가측 공간 $ (U,\mathcal{F}) $에서 실수 보렐 공간 $ (\mathbb{R},\mathcal{B}(\mathbb{R})) $으로 가는 가측 함수
#### $ \Rrightarrow{} f : (U,\mathcal{F}) \rightarrow{} (\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
#### $ \Leftrightarrow{} f^{-1}(B) \in{} \mathcal{F} \;\;\;\; \forall{} B \in{} \mathcal{B}(\mathbb{R}) $

<b></b>

> ## **(Theorem 0-4) Measurability Criterion**

#### 가측 사상의 가측성은 공역의 모든 가측 집합을 직접 확인하지 않고, 해당 시그마-대수의 생성족에 대한 역상만으로 판정할 수 있다.

#### **(1)** 공역의 시그마-대수를 생성하는 집합족의 역상만으로 가측성을 판정할 수 있다.
#### $ \Rrightarrow{} \mathcal{G} = \sigma{}(\mathcal{C}) $
#### $ \Rrightarrow{} f : (U,\mathcal{F}) \rightarrow{} (W,\mathcal{G}) \, \text{ is measurable} \;\; \Leftrightarrow{} \;\; f^{-1}(C) \in{} \mathcal{F} \;\;\;\; \forall{} C \in{} \mathcal{C} $

##### **(`EX`)** 유한 집합 위에서 공역의 생성족 하나만 확인하여 가측성을 판정하고자 할 경우
##### $ \hspace{0.15cm} \text{let } \, U=\{1,2,3,4\},\;\; \mathcal{F}=\{\emptyset{},\{1,2\},\{3,4\},U\} $
##### $ \hspace{0.15cm} \text{and } \, W=\{a,b\},\;\; \mathcal{C}=\{\{a\}\},\;\; \mathcal{G}=\sigma{}(\mathcal{C})=\{\emptyset{},\{a\},\{b\},W\} $
##### $ \hspace{0.15cm} \text{and } \, f(1)=a,\;\; f(2)=a,\;\; f(3)=b,\;\; f(4)=b $
##### $ \hspace{0.45cm} f^{-1}(\{a\})=\{1,2\}\in{}\mathcal{F} $
##### $ \hspace{0.45cm} \therefore{} f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}) \, \text{ is measurable} $

#### **(2)** 실수값 함수는 집합족 $ \{(-\infty{},a] : a \in{} \mathbb{R}\} $의 역상만으로 가측성을 판정할 수 있다.
#### $ \Rrightarrow{} \mathcal{B}(\mathbb{R})=\sigma{}(\{(-\infty{},a] : a \in{} \mathbb{R}\}) $
#### $ \Rrightarrow{} f:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
#### $ \Leftrightarrow{} f^{-1}((-\infty{},a]) \in{} \mathcal{F} \;\;\;\; \forall{} a \in{} \mathbb{R} $
#### $ \hspace{0.45cm} \text{where } \, f^{-1}((-\infty{},a])=\{u\in{}U:f(u)\leq{}a\} $

##### **(`EX`)** 실수값 함수 $ f(u)=u^{2} $의 가측성을 판정하고자 할 경우
##### $ \hspace{0.15cm} \text{let } \, f:(\mathbb{R},\mathcal{B}(\mathbb{R}))\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})),\;\; f(u)=u^{2} $
##### $ \hspace{0.45cm} f^{-1}((-\infty{},a])=\{u\in{}\mathbb{R}:u^{2}\leq{}a\} $
##### $ \hspace{0.45cm} \text{if } \, a<0 \;\; \Rightarrow{} \;\; f^{-1}((-\infty{},a])=\emptyset{}\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \text{if } \, a\geq{}0 \;\; \Rightarrow{} \;\; f^{-1}((-\infty{},a])=[-\sqrt{a},\sqrt{a}]\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \therefore{} f:(\mathbb{R},\mathcal{B}(\mathbb{R}))\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $

<b></b>

> ## **사상이 생성하는 시그마-대수(Sigma-Algebra Generated by Mapping)**

#### **(1) 정의** : 함수 $ f:U\rightarrow{}W $와 공역의 시그마-대수 $ \mathcal{G} $가 존재할 때, $ f $의 역상으로 구성되는 $ U $ 위의 시그마-대수
#### $ \Rrightarrow{} \sigma{}(f):=f^{-1}(\mathcal{G})=\{f^{-1}(B):B\in{}\mathcal{G}\} $

##### **(`PLUS`)** $ \sigma{}(f) $는 함수 $ f $를 가측 함수로 만드는 $ U $ 위의 가장 작은 시그마-대수임

<b></b>

> ## **기초 가측 함수(Elementary Measurable Functions)**

#### **(1) 지시 함수(Indicator Function)** : 가측 집합 $ A \in{} \mathcal{F} $에 대해 원소가 $ A $에 포함되는지 여부를 $ 0 $ 또는 $ 1 $로 나타내는 함수
#### $ \Rrightarrow{} \mathbf{1}_{A}(u):=\begin{cases}1,&u\in{}A\\0,&u\notin{}A\end{cases} $
#### $ \hspace{0.45cm} \mathbf{1}_{A}:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $

#### **(2) 단순 함수(Simple Function)** : 유한 개의 실수값만을 갖는 가측 함수
#### $ \Rrightarrow{} \phi{}(u):=\displaystyle{} \sum_{k=1}^{m} a_{k}\mathbf{1}_{A_{k}}(u) $
#### $ \hspace{0.45cm} \text{where } \, A_{i}\cap{}A_{j}=\emptyset{} \;\; (i\neq{}j),\;\; a_{k}\in{}\mathbb{R},\;\; A_{k}\in{}\mathcal{F},\;\; k=1,\cdots{},m $

<b></b>

> ## **(Theorem 0-5) Stability of Measurable Mapping**

#### 가측 함수는 유한 연산, 점별극한, 합성에 대해 안정적이다.

#### **(1)** 두 실수값 가측 함수로 만든 선형결합, 곱, 절댓값, 최댓값 및 최솟값은 다시 가측 함수이다.
#### $ \Rrightarrow{} f,g \, \text{ are measurable} \; \Rightarrow{} \; \big[{} af+bg,\;\; fg,\;\; |f|,\;\; \max{}\{f,g\},\;\; \min{}\{f,g\} \big]{} \, \text{ are measurable} $
#### $ \hspace{0.45cm} \text{where } \, a,b\in{}\mathbb{R} $

#### **(2)** 분모가 $ 0 $이 되지 않는 경우 두 실수값 가측 함수의 몫은 다시 가측 함수이다.
#### $ \Rrightarrow{} f,g \, \text{ are measurable},\;\; g(u)\neq{}0 \;\; \forall{}u\in{}U \; \Rightarrow{} \; \frac{f}{g} \, \text{ is measurable} $

#### **(3)** 가측 함수열이 점별 수렴하면 그 극한 함수도 가측 함수이다.
#### $ \Rrightarrow{} f_{n} \, \text{ are measurable} \;\; f_{n}(u)\rightarrow{}f(u) \; \Rightarrow{} \; f:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
#### $ \hspace{0.45cm} \text{where } \, f_{n}:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\; n = 1, 2, \cdots{}, \;\; \forall{}u\in{}U $

#### **(4)** 두 가측 사상의 합성은 다시 가측 사상이다.
#### $ \Rrightarrow{} f,g \, \text{ are measurable} \; \Rightarrow{} \; g\circ{}f:(U,\mathcal{F})\rightarrow{}(Z,\mathcal{H}) \, \text{ is measurable} $
#### $ \hspace{0.45cm} \text{where } \, f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}),\;\; g:(W,\mathcal{G})\rightarrow{}(Z,\mathcal{H}) $
