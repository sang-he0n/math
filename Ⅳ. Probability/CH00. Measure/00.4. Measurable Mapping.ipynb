# CH00.4. **Measurable Mapping**

> ## **가측 사상(Measurable Mapping)**

#### **(1) 정의** : 두 가측 공간 사이의 함수 $ f $에 대해, 공역의 모든 가측 집합 $ B $의 역상 $ f^{-1}(B) $가 정의역의 시그마-대수에 포함되는 사상(함수) $ \, f $
#### $ \Rrightarrow{} f : (U,\mathcal{F}) \rightarrow{} (W,\mathcal{G}) \, \text{ is measurable} \;\; \Leftrightarrow{} \;\; f^{-1}(B) \in{} \mathcal{F} \;\;\;\; \forall{} B \in{} \mathcal{G} $
#### $ \hspace{0.45cm} \text{where } \, \mathcal{F} \, \text{ is a } \sigma{}\text{-algebra on } U, \;\; \mathcal{G} \, \text{ is a } \sigma{}\text{-algebra on } W $

#### **(2) 종류** : 
##### $ \hspace{0.15cm} $ ① 실수(Real-Valued) 가측 사상 : 가측 공간 $ (U,\mathcal{F}) $에서 실수 보렐 공간 $ (\mathbb{R},\mathcal{B}(\mathbb{R})) $으로 가는 가측 사상
##### $ \hspace{0.45cm} \Rrightarrow{} f : (U,\mathcal{F}) \rightarrow{} (\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable } \;\; \Leftrightarrow{} \;\; f^{-1}(B)\in{}\mathcal{F} \;\;\;\; \forall{}B\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.15cm} $ ② 보렐(Borel) 가측 사상 : 정의역과 공역이 실수공간의 보렐 시그마-대수를 갖는 가측 사상
##### $ \hspace{0.45cm} \Rrightarrow{} f : (U,\mathcal{B}(U)) \rightarrow{} (W,\mathcal{B}(W)) \, \text{ is Borel measurable} \;\; \Leftrightarrow{} \;\; f^{-1}(B)\in{}\mathcal{B}(U) \;\;\;\; \forall{}B\in{}\mathcal{B}(W) $

> ## (Theorem 0-n) **Measurability Criterion**

#### 가측 사상의 가측성은 공역의 모든 가측 집합을 직접 확인하지 않고, 생성족 또는 반직선의 역상으로 판정할 수 있다.

#### **(1) 생성족 판정(Generator Criterion)** : 공역의 시그마-대수를 생성하는 집합족의 역상만으로 가측성을 판정할 수 있다.
#### $ \Rrightarrow{} \mathcal{G}=\sigma{}(\mathcal{C}), \;\; f^{-1}(C)\in{}\mathcal{F}, \;\;\;\; (\forall{}C\in{}\mathcal{C})  \; \Rightarrow{} \; f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}) \, \text{ is measurable} $

##### **(`EX`)** 유한 집합 위에서 공역의 생성족 하나만 확인하여 가측성을 판정하고자 할 경우
##### $ \hspace{0.15cm} \text{let } \, U=\{1,2,3,4\}, \;\; \mathcal{F}=\{\varnothing{},\{1,2\},\{3,4\},U\} $
##### $ \hspace{0.15cm} \text{and } \, W=\{a,b\}, \;\; \mathcal{C}=\{\{a\}\}, \;\; \mathcal{G}=\sigma{}(\mathcal{C})=\{\varnothing{},\{a\},\{b\},W\} $
##### $ \hspace{0.15cm} \text{and } \, f(1)=a,\;\; f(2)=a,\;\; f(3)=b,\;\; f(4)=b $
##### $ \hspace{0.45cm} f^{-1}(\{a\})=\{1,2\}\in{}\mathcal{F} $
##### $ \hspace{0.45cm} \therefore{} f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G})  \, \text{ is measurable} $

#### **(2) 실수값 함수 판정(Real-Valued Criterion)** 실수값 함수의 가측성은 반직선의 역상만으로 판정할 수 있다.
#### $ \Rrightarrow{} \mathcal{B}(\mathbb{R})=\sigma{}\left(\{(-\infty{},a]:a\in{}\mathbb{R}\}\right), \;\; f^{-1}((-\infty{},a])\in{}\mathcal{F} \;\;\;\; \forall{}a\in{}\mathbb{R} \;\; \Rightarrow{} \;\; f:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
##### $ \hspace{0.45cm} \text{where } \, f^{-1}((-\infty{},a])=\{u\in{}U:f(u)\le{}a\} $

##### **(`PLUS`) 반직선(Half-line; Ray)** 실수선 위에서 기준값 $ a $를 기준으로 한쪽 방향으로 무한히 뻗는 구간
##### $ \hspace{0.15cm} \Rrightarrow{} (-\infty{},a]=\{x\in{}\mathbb{R}:x\le{}a\}, \;\; (-\infty{},a)=\{x\in{}\mathbb{R}:x<a\}, \;\; \cdots{} $

##### **(`EX`)** 실수값 함수의 가측성을 반직선의 역상으로 판정하고자 할 경우
##### $ \hspace{0.15cm} \text{let } \, f:(\mathbb{R},\mathcal{B}(\mathbb{R}))\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\; f(u)=u^{2} $
##### $ \hspace{0.45cm} f^{-1}((-\infty{},a])=\{u\in{}\mathbb{R}:u^{2}\le{}a\} $
##### $ \hspace{0.45cm} \text{if } \, a<0 \;\; \Rightarrow{} \;\; f^{-1}((-\infty{},a])=\varnothing{}\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \text{if } \, a\ge{}0 \;\; \Rightarrow{} \;\; f^{-1}((-\infty{},a])=[-\sqrt{a},\sqrt{a}]\in{}\mathcal{B}(\mathbb{R}) $
##### $ \hspace{0.45cm} \therefore{} f:(\mathbb{R},\mathcal{B}(\mathbb{R}))\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $

> ## **기초 가측 함수(Elementary Measurable Functions)**

#### **(1) 정의** : 어떠한 부분 집합 $ A \subseteq{} U $에 대해, 원소가 $ A $에 포함되는지 여부를 $ 0 $ 또는 $ 1 $로 나타내는 함수
#### $ \Rrightarrow{} \mathbf{1}_{A}:U\rightarrow{}\{0,1\} \; \Leftrightarrow{} \; \mathbf{1}_{A}(u):= \begin{cases} 1, & u\in{}A \\ 0, & u\notin{}A \end{cases} $

<b></b>

> ## (Theorem 0-n) **Stability of Measurable Mapping**

#### 가측 함수는 유한 연산, 점별극한, 합성에 대해 안정적이다.

#### **(1)** 두 실수값 가측 함수로 만든 선형결합, 곱, 절댓값, 최댓값, 최솟값은 다시 가측 함수이다.
#### $ \Rrightarrow{} f,g \, \text{ are measurable} \; \Rightarrow{} \; \big[{} af+bg, \;\; fg, \;\; |f|, \;\; \max{}\{f,g\}, \;\; \min{}\{f,g\} \big]{} \, \text{ are measurable} $
#### $ \hspace{0.45cm} \text{where } \, f,g:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\;  a,b\in{}\mathbb{R} $

#### **(2)** 분모가 $ 0 $이 되지 않는 경우 두 실수값 가측 함수의 몫은 다시 가측 함수이다.
#### $ \Rrightarrow{} f,g \, \text{ are measurable} \;\; g(u)\neq{}0 \; \Rightarrow{} \; \frac{f}{g}:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
#### $ \hspace{0.45cm} \text{where } \, f,g:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\; \forall{}u\in{}U $

#### **(3)** 가측 함수열의 상한, 하한, 상극한, 하극한은 다시 가측 함수이다.
#### $ \Rrightarrow{} \displaystyle{} f_{n} \, \text{ are measurable} \; \Rightarrow{} \; \big[{} \sup{} f_{n}, \;\; \inf{} f_{n}, \;\; \limsup_{n\to{}\infty{}}f_{n}, \;\; \liminf_{n\to{}\infty{}}f_{n} \big]{} \, \text{ are measurable} $
#### $ \hspace{0.45cm} \text{where } \, f_{n}:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\; n = 1, 2, \cdots{} $

#### **(4)** 가측 함수열이 점별 수렴하면 그 극한 함수도 가측 함수이다.
#### $ \Rrightarrow{} f_{n} \, \text{ are measurable} \;\; f_{n}(u)\rightarrow{}f(u) \; \Rightarrow{} \; f:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})) \, \text{ is measurable} $
#### $ \hspace{0.45cm} \text{where } \, f_{n}:(U,\mathcal{F})\rightarrow{}(\mathbb{R},\mathcal{B}(\mathbb{R})), \;\; n = 1, 2, \cdots{}, \;\; \forall{}u\in{}U $

#### **(5)** 두 가측 사상의 합성은 다시 가측 사상이다.
#### $ \Rrightarrow{} f, g \, \text{ are measurable} \; \Rightarrow{} \; g\circ{}f:(U,\mathcal{F})\rightarrow{}(Z,\mathcal{H}) \, \text{ is measurable} $
#### $ \hspace{0.45cm} \text{where } \, f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}), \;\; g:(W,\mathcal{G})\rightarrow{}(Z,\mathcal{H}) $

> ## **사상이 생성하는 시그마-대수(Sigma-Algebra Generated by Mapping)**

#### **(1) 정의** : 함수 $ f:U\rightarrow{}W $와 공역의 시그마-대수 $ \mathcal{G} $가 존재할 때, $ f $의 역상들로 생성되는 $ U $ 위의 시그마-대수
#### $ \Rrightarrow{} \sigma{}(f):=f^{-1}(\mathcal{G})=\{f^{-1}(B):B\in{}\mathcal{G}\} $
