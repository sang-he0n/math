# CH00.6. **Lebesgue Integration**

> ## **르베그 적분(Lebesgue Integral)**

#### **(1) 정의** : 측도 공간 $ (U,\mathcal{F},\mu{}) $에서 가측 함수 $ f $의 값을 측도 $ \mu{} $에 따라 집계하도록 정의한 적분
#### $ = $ 가측 집합의 측도값에서 출발하여 지시 함수, 단순 함수, 비음수 가측 함수, 일반 가측 함수 순으로 확장하여 정의한 적분
#### $ \Rrightarrow{} \displaystyle{} \int_{U} f\,d\mu{} $

#### **(2) 가측 집합 위의 적분** : 가측 집합 $ A \in{} \mathcal{F} $ 위의 적분은 지시 함수를 이용하여 정의함
#### $ \Rrightarrow{} \displaystyle{} \int_{A} f\,d\mu{} := \int_{U} f\mathbf{1}_{A}\,d\mu{} $

<b></b>

> ## **르베그 적분의 구성(Construction of Lebesgue Integral)**

#### **(1) 지시 함수의 적분(Integral of Indicator Function)** : 가측 집합 $ A \in{} \mathcal{F} $의 지시 함수에 대한 적분을 집합 $ A $의 측도로 정의함
#### $ \Rrightarrow{} \displaystyle{} \int_{U} \mathbf{1}_{A}\,d\mu{} := \mu{}(A) $

##### **(`PLUS`)** 르베그 적분은 가측 집합의 크기를 측정하는 측도에서 출발하여 가측 함수의 값을 측정하는 연산으로 확장됨

#### **(2) 단순 함수의 적분(Integral of Simple Function)** : 유한 개의 지시 함수로 구성된 비음수 단순 함수의 적분을 각 가측 집합의 측도에 함수값을 곱한 값의 합으로 정의함
#### $ \Rrightarrow{} \phi{}(u) := \displaystyle{} \sum_{k=1}^{m} a_{k}\mathbf{1}_{A_{k}}(u) $
#### $ \Rrightarrow{} \displaystyle{} \int_{U} \phi{}\,d\mu{} := \sum_{k=1}^{m} a_{k}\mu{}(A_{k}) $
#### $ \hspace{0.45cm} \text{where } \, a_{k} \geq{} 0,\;\; A_{k} \in{} \mathcal{F},\;\; A_{i}\cap{}A_{j}=\emptyset{} \;\; \forall{} i\neq{}j $

##### **(`PLUS`)** 모든 비음수 가측 함수는 아래에서 단조 증가하는 비음수 단순 함수열로 근사할 수 있음
##### $ \hspace{0.15cm} \Rrightarrow{} 0 \leq{} \phi_{1} \leq{} \phi_{2} \leq{} \cdots{} \leq{} f,\;\; \phi_{n}(u)\uparrow{}f(u) $

#### **(3) 비음수 가측 함수의 적분(Integral of Nonnegative Measurable Function)** : 비음수 가측 함수 $ f $보다 작거나 같은 모든 비음수 단순 함수의 적분 중 상한으로 정의함
#### $ = $ 비음수 가측 함수 $ f $를 아래에서 근사하는 단순 함수들의 적분값을 최대한 $ f $에 가깝게 확장한 값
#### $ \Rrightarrow{} \displaystyle{} \int_{U} f\,d\mu{} := \sup{}\left\{\int_{U}\phi{}\,d\mu{} : 0\leq{}\phi{}\leq{}f,\;\; \phi{} \text{ is nonnegative simple measurable}\right\} $
#### $ \hspace{0.45cm} \text{where } \, f:U\rightarrow{}[0,\infty{}] \text{ is measurable} $

##### **(`PLUS`)** 비음수 가측 함수의 르베그 적분은 무한한 값을 가질 수 있음
##### $ \hspace{0.15cm} \Rrightarrow{} \displaystyle{} \int_{U} f\,d\mu{} \in{} [0,\infty{}] $

#### **(4) 양의 부분과 음의 부분(Positive Part and Negative Part)** : 실수값 가측 함수 $ f $를 각각 비음수 값을 갖는 양의 부분 $ f^{+} $와 음의 부분 $ f^{-} $로 분해함
##### $ \hspace{0.15cm} $ ① 양의 부분(Positive Part)
##### $ \hspace{0.45cm} f^{+}(u) := \max{}\{f(u),0\} $
##### $ \hspace{0.15cm} $ ② 음의 부분(Negative Part)
##### $ \hspace{0.45cm} f^{-}(u) := \max{}\{-f(u),0\} $
#### $ \Rrightarrow{} f = f^{+}-f^{-},\;\; |f| = f^{+}+f^{-} $

#### **(5) 일반 가측 함수의 적분(Integral of General Measurable Function)** : 실수값 가측 함수 $ f $의 적분을 양의 부분과 음의 부분의 적분 차이로 정의함
#### $ \Rrightarrow{} \displaystyle{} \int_{U} f\,d\mu{} := \int_{U} f^{+}\,d\mu{} - \int_{U} f^{-}\,d\mu{} $
#### $ \hspace{0.45cm} \text{where } \, \displaystyle{} \int_{U} f^{+}\,d\mu{} < \infty{} \;\; \text{or} \;\; \int_{U} f^{-}\,d\mu{} < \infty{} $

##### **(`PLUS`)** 두 적분이 모두 무한한 경우 $ \infty{}-\infty{} $가 정의되지 않으므로 $ f $의 르베그 적분 역시 정의되지 않음

<b></b>

> ## **적분 가능성(Integrability)**

#### **(1) 정의** : 실수값 가측 함수 $ f $의 절댓값에 대한 르베그 적분이 유한한 경우 $ f $를 측도 $ \mu{} $에 대해 적분 가능한 함수라고 함
#### $ \Rrightarrow{} f \text{ is } \mu{}\text{-integrable} \;\; \Leftrightarrow{} \;\; \displaystyle{} \int_{U}|f|\,d\mu{} < \infty{} $

#### **(2) 적분 가능 조건** : 실수값 가측 함수 $ f $가 적분 가능할 필요충분조건은 양의 부분과 음의 부분의 적분이 모두 유한한 것임
#### $ \Rrightarrow{} \displaystyle{} \int_{U}|f|\,d\mu{} < \infty{} $
#### $ \Leftrightarrow{} \displaystyle{} \int_{U}f^{+}\,d\mu{} < \infty{},\;\; \int_{U}f^{-}\,d\mu{} < \infty{} $

##### **(`PLUS`)** 측도 $ \mu{} $에 대해 적분 가능한 함수들의 공간을 $ L^{1}(\mu{}) $로 표현함
##### $ \hspace{0.15cm} \Rrightarrow{} L^{1}(\mu{}) := \left\{f : \displaystyle{} \int_{U}|f|\,d\mu{} < \infty{}\right\} $

<b></b>

> ## **(Theorem 0-6) Basic Properties of Lebesgue Integral**

#### 르베그 적분은 적분 가능한 함수에 대해 다음의 기본 성질을 만족한다.

#### **(1)** 두 적분 가능한 함수의 선형결합에 대한 적분은 각 함수의 적분에 동일한 계수를 적용한 선형결합과 동일하다.
#### $ \Rrightarrow{} \displaystyle{} \int_{U}(af+bg)\,d\mu{} = a\int_{U}f\,d\mu{} + b\int_{U}g\,d\mu{} $
#### $ \hspace{0.45cm} \text{where } \, f,g \in{} L^{1}(\mu{}),\;\; a,b \in{} \mathbb{R} $

#### **(2)** 두 적분 가능한 함수가 거의 어디서나 대소관계를 만족하면 해당 함수들의 적분값 역시 동일한 대소관계를 만족한다.
#### $ \Rrightarrow{} f \leq{} g \;\; \mu{}\text{-a.e.} \;\; \Rightarrow{} \;\; \displaystyle{} \int_{U}f\,d\mu{} \leq{} \int_{U}g\,d\mu{} $

#### **(3)** 적분 가능한 함수의 적분값의 절댓값은 해당 함수의 절댓값을 적분한 값보다 작거나 같다.
#### $ \Rrightarrow{} \displaystyle{} \left|\int_{U}f\,d\mu{}\right| \leq{} \int_{U}|f|\,d\mu{} $

#### **(4)** 두 적분 가능한 함수가 거의 어디서나 동일하면 두 함수의 르베그 적분값 역시 동일하다.
#### $ \Rrightarrow{} f=g \;\; \mu{}\text{-a.e.} \;\; \Rightarrow{} \;\; \displaystyle{} \int_{U} f\,d\mu{} = \int_{U}g\,d\mu{} $

<b></b>

> ## **(Theorem 0-7) Change of Variables by Pushforward**

#### 측도 $ \mu{} $를 가측 사상 $ f $를 통해 푸시포워드한 측도 $ f_{\#}\mu{} $에 대한 함수 $ g $의 적분은 원래 측도 $ \mu{} $에 대한 합성 함수 $ g\circ{}f $의 적분과 같다.
#### $ \Rrightarrow{} \displaystyle{} \int_{W}g\,d(f_{\#}\mu{}) = \int_{U}(g\circ{}f)\,d\mu{} $
#### $ \hspace{0.45cm} \text{where } \, f:(U,\mathcal{F})\rightarrow{}(W,\mathcal{G}) \text{ is measurable},\;\; g \in{} L^{1}(f_{\#}\mu{}) $

<b></b>

> ## **(Theorem 0-8) Monotone Convergence Theorem**

#### 비음수 가측 함수열 $ \{f_{n}\} $이 점별로 단조 증가하여 $ f $로 수렴하면 함수열의 적분 역시 $ f $의 적분으로 단조 수렴한다.
#### $ \Rrightarrow{} 0\leq{}f_{1}\leq{}f_{2}\leq{}\cdots{},\;\; f_{n}(u)\uparrow{}f(u) $
#### $ \Rrightarrow{} \displaystyle{} \lim_{n\rightarrow{}\infty{}}\int_{U}f_{n}\,d\mu{} = \int_{U}f\,d\mu{} $

<b></b>

> ## **(Theorem 0-9) Fatou's Lemma**

#### 비음수 가측 함수열에서 함수열의 하극한을 적분한 값은 각 함수의 적분값들의 하극한보다 작거나 같다.
#### $ \Rrightarrow{} \displaystyle{} \int_{U}\liminf_{n\rightarrow{}\infty{}}f_{n}\,d\mu{} \leq{} \liminf_{n\rightarrow{}\infty{}}\int_{U}f_{n}\,d\mu{} $
#### $ \hspace{0.45cm} \text{where } \, f_{n}\geq{}0 \;\; \forall{}n $

##### **(`PLUS`)** 함수열과 적분의 극한을 직접 교환할 수 없는 경우에도 적분값의 하한을 비교하는 데 사용할 수 있음

<b></b>

> ## **(Theorem 0-10) Dominated Convergence Theorem**

#### 가측 함수열 $ \{f_{n}\} $이 가측 함수 $ f $로 거의 어디서나 수렴하고 모든 $ f_{n} $이 하나의 적분 가능한 함수 $ g $에 의해 지배되면 극한과 적분의 순서를 교환할 수 있다.
#### $ \Rrightarrow{} f_{n}\rightarrow{}f \;\; \mu{}\text{-a.e.},\;\; |f_{n}|\leq{}g \;\; \mu{}\text{-a.e.},\;\; g\in{}L^{1}(\mu{}) $
#### $ \Rrightarrow{} \displaystyle{} \lim_{n\rightarrow{}\infty{}}\int_{U}f_{n}\,d\mu{} = \int_{U}f\,d\mu{} $

##### **(`PLUS`)** 위 조건에서는 $ L^{1} $ 수렴 역시 성립함
##### $ \hspace{0.15cm} \Rrightarrow{} \displaystyle{} \lim_{n\rightarrow{}\infty{}}\int_{U}|f_{n}-f|\,d\mu{} = 0 $

<b></b>

> ## **곱 측도에 대한 적분(Product Integration)**

#### **(1) 정의** : 두 측도 공간 $ (U,\mathcal{F},\mu{}) $, $ (W,\mathcal{G},\nu{}) $의 곱 측도 $ \mu{}\otimes{}\nu{} $에 대해 곱 공간 $ U\times{}W $ 위의 가측 함수를 적분하는 방법
#### $ \Rrightarrow{} \displaystyle{} \int_{U\times{}W}f\,d(\mu{}\otimes{}\nu{}) $

<b></b>

> ## **(Theorem 0-11) Tonelli's Theorem**

#### 두 시그마-유한 측도 공간의 곱 공간에서 비음수 가측 함수의 곱 측도에 대한 적분은 두 반복적분과 같으며 적분 순서를 교환할 수 있다.
#### $ \Rrightarrow{} \displaystyle{} \int_{U\times{}W}f\,d(\mu{}\otimes{}\nu{}) $
#### $ \hspace{0.45cm} = \displaystyle{} \int_{U}\left[\int_{W}f(u,w)\,d\nu{}(w)\right]d\mu{}(u) $
#### $ \hspace{0.45cm} = \displaystyle{} \int_{W}\left[\int_{U}f(u,w)\,d\mu{}(u)\right]d\nu{}(w) $
#### $ \hspace{0.45cm} \text{where } \, f\geq{}0,\;\; \mu{},\nu{} \text{ are } \sigma{}\text{-finite} $

<b></b>

> ## **(Theorem 0-12) Fubini's Theorem**

#### 두 시그마-유한 측도 공간의 곱 공간에서 함수 $ f $가 절대 적분 가능하면 곱 측도에 대한 적분은 두 반복적분과 같으며 적분 순서를 교환할 수 있다.
#### $ \Rrightarrow{} \displaystyle{} \int_{U\times{}W}|f|\,d(\mu{}\otimes{}\nu{}) < \infty{} $
#### $ \Rrightarrow{} \displaystyle{} \int_{U\times{}W}f\,d(\mu{}\otimes{}\nu{}) $
#### $ \hspace{0.45cm} = \displaystyle{} \int_{U}\left[\int_{W}f(u,w)\,d\nu{}(w)\right]d\mu{}(u) $
#### $ \hspace{0.45cm} = \displaystyle{} \int_{W}\left[\int_{U}f(u,w)\,d\mu{}(u)\right]d\nu{}(w) $
#### $ \hspace{0.45cm} \text{where } \, \mu{},\nu{} \text{ are } \sigma{}\text{-finite} $
